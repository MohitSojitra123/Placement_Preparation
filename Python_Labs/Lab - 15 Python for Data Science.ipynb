## Given an image file:

```python
from PIL import Image #Python Imagining Library (PIL)
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

### 01) Read the image using the PIL library.

```python
img1 = Image.open("logo.png")
```

```python
plt.imshow(img1)
plt.show()
```

```python
plt.imshow(img1)
plt.axis("off")
plt.show()
```

### 02) Convert the image into a NumPy array.

```python
img1_array = np.array(img1)
```

```python
print(img1_array.shape)
```

### 03) Create a copy of the image array.

```python
img_copy = img1_array.copy()
```

### 04) Extract a sub-image starting from the 100th row and 100th column, selecting every 20th row and column.

```python
subimg = img_copy[100::20, 100::20]
```

### 05) Display the processed image using Matplotlib.

```python
plt.imshow(subimg)
plt.axis("off")
plt.show()
```
