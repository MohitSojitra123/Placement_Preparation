### 16) Text Formatter
**Problem Definition**
Accept a paragraph and remove extra spaces, convert to Title Case and count words.

**Sample**
```
Input:
   python   programming    is   interesting
Output:
Formatted Text : Python Programming Is Interesting
Total Words : 4
```

````python
s = input("Enter paragraph: ")

s = " ".join(s.split())
s = s.title()

w = len(s.split())

print("Formatted Text :", s)
print("Total Words :", w)
````

### 17) Email Validator

**Problem Definition**

Validate an email using string methods.
A valid email:
- contains exactly one '@'
- contains at least one '.'
- does not start or end with '@' or '.'
- contains no spaces.

**Sample**
```
Input: student@gmail.com
Output: Valid Email
```


````python
e = input("Enter email: ")

if e.count("@") == 1 and "." in e and not e.startswith(("@", ".")) and not e.endswith(("@", ".")) and " " not in e:
    print("Valid Email")
else:
    print("Invalid Email")
````

### 18) Username Generator

**Problem Definition**

Generate username from full name and birth year
Generate a username by:
- removing extra spaces,
- converting the name to lowercase,
- replacing spaces with underscores,
- appending the last two digits of the birth year.

**Sample**

```
Input:
Name : Rahul Kumar Sharma
Birth Year : 2004

Output:
Username : rahul_kumar_sharma04
```

````python
n = input("Name : ")
y = input("Birth Year : ")

n = "_".join(n.split()).lower()

u = n + y[-2:]

print("Username :", u)
````

### 19) Sentence Statistics

**Problem Definition**

Display number of words,vowels,consonants,digits,special chars.

**Sample**

```
Input:
Python3 is Awesome!

Output:
Words : 3
Vowels : 6
Consonants : 9
Digits : 1
Special Characters : 1
```

````python
s = input("Enter text: ")

w = len(s.split())
v = 0
c = 0
d = 0
sp = 0

for i in s:
    if i.lower() in "aeiou":
        v += 1
    elif i.isalpha():
        c += 1
    elif i.isdigit():
        d += 1
    elif i != " ":
        sp += 1

print("Words :", w)
print("Vowels :", v)
print("Consonants :", c)
print("Digits :", d)
print("Special Characters :", sp)
````

### 20) Word Replacement Utility

**Problem Definition**

Accept a sentence, a word to search, and a replacement word. Replace only if the word exists; otherwise display an appropriate message.

**Sample**

```
Input:
Sentence : Python is easy to learn.
Search : easy
Replace : powerful

Output:
Updated Sentence : Python is powerful to learn.
```

````python
s = input("Sentence : ")
a = input("Search : ")
b = input("Replace : ")

if a in s:
    s = s.replace(a, b)
    print("Updated Sentence :", s)
else:
    print("Word not found")
````

### 22) Product Code Analyzer

**Problem Definition**

A product code is in the format ABC-1234-XY. Write a program to:
- Separate all three parts.
- Verify that:
        - first part contains only alphabets,
        - second part contains only digits,
        - third part contains only uppercase letters.

**Sample**

```
Input:
ABC-1234-XY

Output:
Code : ABC
Number : 1234
Category : XY
Valid Product Code
```

````python
s = input("Enter code: ")

p = s.split("-")

if len(p) == 3:
    a = p[0]
    b = p[1]
    c = p[2]

    print("Code :", a)
    print("Number :", b)
    print("Category :", c)

    if a.isalpha() and b.isdigit() and c.isupper() and c.isalpha():
        print("Valid Product Code")
    else:
        print("Invalid Product Code")
else:
    
    print("Invalid Product Code")
````
