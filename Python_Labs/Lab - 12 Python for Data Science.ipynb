### 1. Concatenate DataFrames Row-wise

Create two DataFrames containing student details from two different groups. Use `pd.concat()` with `axis=0` to concatenate the DataFrames row-wise. Display the original DataFrames and the resulting DataFrame.


```python
import pandas as pd


df1 = pd.DataFrame({
    'RollNo': [101, 102, 103],
    'Name': ['Ravi', 'Amit', 'Neha'],
    'Marks': [85, 78, 90]
})

df2 = pd.DataFrame({
    'RollNo': [104, 105, 106],
    'Name': ['Raj', 'Pooja', 'Karan'],
    'Marks': [75, 88, 82]
})


df3 = pd.concat([df1, df2], axis=0)


print("Group 1:")
print(df1)

print("\nGroup 2:")
print(df2)


print("\nCombined DataFrame:")
print(df3)
```

### 6. Join Two DataFrames using Index

Create two DataFrames containing student information and marks, with `RollNo` as the index. Use the `join()` method to combine the DataFrames based on their index. Display the resulting DataFrame.

### 7. Join DataFrames with Different Index Values

Create two DataFrames containing student details and attendance information with partially matching indexes. Demonstrate `join()` using the default left join and then using `how='inner'`. Compare the results.


```python

df1 = pd.DataFrame({
    'Name': ['Ravi', 'Amit', 'Neha', 'Pooja'],
    'Course': ['BCA', 'BCA', 'BCA', 'BCA']
}, index=[101, 102, 103, 104])

df1.index.name = 'RollNo'


df2 = pd.DataFrame({
    'Attendance': [85, 90, 78]
}, index=[101, 103, 105])

df2.index.name = 'RollNo'

left_result = df1.join(df2)

inner_result = df1.join(df2, how='inner')

print("Left Join:")
print(left_result)

print("\nInner Join:")
print(inner_result)
```

### 8. Merge DataFrames using a Common Column

Create two DataFrames containing student details and examination marks. Both DataFrames should contain a common column named `RollNo`. Use `pd.merge()` to combine them based on `RollNo` and display the result.


```python

df1 = pd.DataFrame({
    'RollNo': [101, 102, 103],
    'Name': ['Ravi', 'Amit', 'Neha'],
    'Course': ['BCA', 'BCA', 'BCA']
})
df2 = pd.DataFrame({
    'RollNo': [101, 102, 103],
    'Marks': [85, 78, 90],
    'Grade': ['A', 'B', 'A']
})

result = pd.merge(df1, df2, on='RollNo')

print(result)
```

### 9. Merge DataFrames using Different Join Types

Create two DataFrames containing student records from two different sources with some common and some non-common `RollNo` values. Use `pd.merge()` to demonstrate `inner`, `left`, `right`, and `outer` joins. Display and compare all four results.


```python

df1 = pd.DataFrame({
    'RollNo': [101, 102, 103, 104],
    'Name': ['Ravi', 'Amit', 'Neha', 'Pooja']
})

df2 = pd.DataFrame({
    'RollNo': [103, 104, 105, 106],
    'Marks': [85, 90, 78, 88]
})


inner = pd.merge(df1, df2, on='RollNo', how='inner')

left = pd.merge(df1, df2, on='RollNo', how='left')

right = pd.merge(df1, df2, on='RollNo', how='right')

outer = pd.merge(df1, df2, on='RollNo', how='outer')


print("Inner Join:")
print(inner)

print("\nLeft Join:")
print(left)

print("\nRight Join:")
print(right)

print("\nOuter Join:")
print(outer)
```

### 10. Combine Student Academic Data using Concat, Join and Merge

Create three DataFrames containing student personal information, semester marks, and attendance information. Use `pd.concat()` to combine two suitable DataFrames, `join()` to combine data using an index, and `pd.merge()` to combine data using a common column. Display the final combined student dataset.


```python
import pandas as pd


student = pd.DataFrame({
    'RollNo': [101, 102, 103],
    'Name': ['Ravi', 'Amit', 'Neha'],
    'Course': ['BCA', 'BCA', 'BCA']
})

marks = pd.DataFrame({
    'RollNo': [101, 102, 103],
    'Marks': [85, 78, 90],
    'Status': ['Pass', 'Pass', 'Pass']
})


attendance = pd.DataFrame({
    'Attendance': [92, 85, 88],
    'Status': ['Good', 'Good', 'Good']
}, index=[101, 102, 103])

attendance.index.name = 'RollNo'


newdf1 = pd.concat([student, marks], axis=1)

print("After Concat:")
print(newdf1)


marks1 = marks.set_index('RollNo')

newdf2 = marks1.join( attendance,how="inner", lsuffix="_marks",rsuffix="_attendance")

print("\nAfter Join:")
print(newdf2)


newdf3 = pd.merge( student, marks, on='RollNo')

newdf3 = pd.merge( newdf3, attendance.reset_index(), on='RollNo' )

print("\nFinal Combined Student Dataset:")
print(newdf3)
```
