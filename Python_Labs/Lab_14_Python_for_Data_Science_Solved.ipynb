### 04) WAP to plot a line chart showing temperature variation during a week. Mark the lowest and highest temperatue.

```python
Days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
Temperature = [20, 32, 31, 33, 34, 43, 36]  # in °C
```

```python
plt.plot(Days, Temperature, marker="o", color="orange")

max_t, min_t = max(Temperature), min(Temperature)
max_d = Days[Temperature.index(max_t)]
min_d = Days[Temperature.index(min_t)]

plt.scatter(max_d, max_t, color="red", s=120, zorder=3, label=f"Highest: {max_t}°C ({max_d})")
plt.scatter(min_d, min_t, color="blue", s=120, zorder=3, label=f"Lowest: {min_t}°C ({min_d})")
plt.annotate("Highest", (max_d, max_t), textcoords="offset points", xytext=(0, 10), ha="center")
plt.annotate("Lowest", (min_d, min_t), textcoords="offset points", xytext=(0, -18), ha="center")

plt.title("Temperature Variation During a Week")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(True)
plt.show()
```

### 05) WAP to create a detailed pie chart representing the monthly expenditure distribution of a family. 
### The chart should display labels, percentage values, title, different colors for each slice, and explode one slice to highlight the highest expense.

```python
Categories = ["Rent", "Food", "Education", "Transport", "Entertainment", "Savings"]
Amount = [15000, 8000, 6000, 3000, 2000, 5000]  # in ₹
```

```python
colors = ["#ff9999", "#66b3ff", "#99ff99", "#ffcc99", "#c2c2f0", "#ffb3e6"]
explode = [0.1 if a == max(Amount) else 0 for a in Amount]   # explode highest expense

plt.figure(figsize=(7, 7))
plt.pie(Amount, labels=Categories, autopct="%1.1f%%", colors=colors,
        explode=explode, startangle=90, shadow=True)
plt.title("Monthly Expenditure Distribution of a Family")
plt.show()
```

### 10) WAP to create a box plot representing the distribution of daily commuting time (in minutes) of employees and identify outliers (if any).

```python
Commuting_Time = [5, 18, 20, 22, 25, 27, 28, 30, 32, 35, 38, 40, 42, 90]  # minutes
```

```python
plt.boxplot(Commuting_Time, vert=False)
plt.title("Daily Commuting Time of Employees")
plt.xlabel("Commuting Time (minutes)")
plt.show()

# Identify outliers using the IQR method
q1, q3 = np.percentile(Commuting_Time, [25, 75])
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
outliers = [t for t in Commuting_Time if t < lower or t > upper]
print("Q1 =", q1, ", Q3 =", q3, ", IQR =", iqr)
print("Outlier bounds:", lower, "to", upper)
print("Outliers:", outliers)
```
