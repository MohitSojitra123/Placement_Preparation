### 14) WAP to extract elements with frequency greater than K.

```python
list1=[1, 2, 2, 1,1,1,3, 2, 4, 5, 6,9,9,96, 7, 4, 8, 9, 9,10, 10, 44]

k=int(input("Enter K : "))

freq_d1={}

for i in list1:
    if i not in freq_d1:
        freq_d1[i]=1
    else:
        freq_d1[i]=freq_d1[i]+1

list2=[]

for keyy,vall in freq_d1.items():
    if(vall>k):
        list2.append(keyy)

print(list2) 
```
