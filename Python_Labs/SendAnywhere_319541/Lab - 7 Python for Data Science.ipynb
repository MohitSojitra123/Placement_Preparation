### 05) WAP to Merge two dictionaries. If a key exists in both dictionaries, add their values.


```
{1: 10, 2: 60, 3: 80, 4: 60}
```

````python
Dict1 = {1: 10, 2: 20, 3: 30}
Dict2 = {2: 40, 3: 50, 4: 60}



for k,v in Dict1.items():
    if k in Dict2 :
        Dict2[k]+=v
    else:
        Dict2[k]=v

print(Dict2)       
````

### 07) WAP to find the keys having maximum and minimum values in a dictionary.

**Sample Input:**
```
{'A': 50, 'B': 90, 'C': 30}
```
**Sample Output:**
```
Max Value Key: B
Min Value Key: C

```

````python
d1={'A': 50, 'B': 90, 'C': 30}

max=d1['A']
min=d1['B']
max_alpha='-'
min_alpha='-'

for k,v in d1.items():
    if v>max :
        max=v
        max_alpha=k
    if v<min :
        min=v
        min_alpha=k

print(max,' -> ',max_alpha)
print(min,' -> ',min_alpha)
````

### 11) WAP to find the student with the highest total marks, given student names and their marks.

**Sample Input:**
```
{'Amit': [70, 80, 90], 'Riya': [85, 75, 80], 'Neha': [60, 70, 65]}
```
**Sample Output:**
```
Topper: Amit
```

````python
d1={'Amit': [70, 80, 90], 'Riya': [85, 75, 80], 'Neha': [60, 70, 65]}


max=0
current_name=""

for k,v in d1.items():
    current_sum=0
    for i in v:
      current_sum+=i

    if current_sum>max:
       max=current_sum
       current_name=k

print("Topper : ",current_name)
````

### 15) WAP to create a dictionary using dictionary comprehension from two lists containing student names and marks. Include only those students in the dictionary whose marks are greater than or equal to 75.

**Sample Input:**
```
students = ['Amit', 'Riya', 'Neha', 'Karan']
marks = [78, 72, 85, 60]
```
**Sample Output:**
```
{'Amit': 78, 'Neha': 85}
```

````python
students = ['Amit', 'Riya', 'Neha', 'Karan']
marks = [78, 72, 85, 60]

d1={students[i]:marks[i] for i in range(0,len(marks),1) if marks[i] >= 75 }

print(d1)
````
