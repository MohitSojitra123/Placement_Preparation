### 1. Create a MultiIndexed DataFrame

Create a DataFrame containing student marks for different Departments and Semesters. Use `set_index()` to create a MultiIndex using `Department` and `Semester`. Display the DataFrame and its index.

```python
import pandas as pd

import numpy as np

data = {
    "RollNo": [101, 102, 103, 104, 105, 106],
    "Name": ["Aarav", "Bhavya", "Chetan", "Dev", "Esha", "Farhan"],
    "Department": ["CSE", "ECE", None, "ME", "ECE", "CSE"],
    "Year": [1, 2, 1, 2, 1, 2],
    "Gender": ["Male", "Female", "Female", "Male", "Male", "Female"],
    "Semester": [3, 4, 3, 4, 3, 4],
    "Marks": [85, np.nan, 65, 88, 75, 58],
    "City": ["Mumbai", "Delhi", "Bangalore", "Chennai", None, "Pune"],
    "Subject": ["Python", "DBMS", "Network", "Java", "Maths", "OS"]
}

students = pd.DataFrame(data)

students.set_index(["Department", "Semester"], inplace=True)

print(students)
print(students.index)
```

### 2. Create a MultiIndex using Multiple Columns

Create a DataFrame containing `Department`, `Year`, `Gender`,`Student`and `Marks`. Set `Department`, `Year` and `Gender` as the MultiIndex and display the resulting DataFrame.

```python
students = pd.DataFrame(data)

students.set_index(["Department", "Year", "Gender"], inplace=True)

print(students)
```

### 3. Read a MultiIndexed DataFrame Directly from CSV

Create a CSV file containing columns `Department`, `Semester`, `Subject`, and `Marks`. Read the CSV file directly as a MultiIndexed DataFrame using `pd.read_csv()` with the appropriate `index_col` parameter. Display the resulting DataFrame.

```python
df = pd.DataFrame(data)


df.to_csv("marks.csv", index=False)
students = pd.read_csv(
    "marks.csv",
    index_col=["Department", "Semester", "Subject"]
)

print(students)
```

### 5. Access Data using MultiIndex

Using a MultiIndexed DataFrame having `Department` and `Semester` as indexes, access all records belonging to one department and one specific department-semester combination.

```python
students = pd.DataFrame(data)

students.set_index(["Department", "Semester"], inplace=True)

print(students)
print(students.loc[('CSE',3)])
```

### 6. Cross Section using xs()

Create a MultiIndexed DataFrame with `Department` and `Semester` as index levels. Use `xs()` to retrieve the records for a particular semester across all departments.

```python
students = pd.DataFrame(data)


students.set_index(["Department", "Semester"], inplace=True)

print(students.xs(4, level="Semester"))
```

### 7. Cross Section from a Specific Level

Using the same MultiIndexed DataFrame, use `xs()` with the `level` parameter to retrieve records for one department. Display the result.

```python
students = pd.DataFrame(data)

students.set_index(["Department", "Semester"], inplace=True)

print(students.xs("CSE", level="Department"))
```

### 8. Cross Section on Multiple Index Levels

Create a MultiIndexed DataFrame with `Department`, `Semester`, and `Subject` as index levels. Use `xs()` to retrieve the records corresponding to a particular combination of Department and Semester.

```python
students = pd.DataFrame(data)

students.set_index( ["Department", "Semester", "Subject"], inplace=True)

print(students)

print(students.xs(("CSE", 3), level=("Department", "Semester")))
```

### 9. Identify Missing Data

Create a student DataFrame containing `Name`, `Department`, `Marks`, `Attendance`, and `City`, with some missing values represented by `NaN` or `None`. Display the DataFrame and identify the missing values using `isna()`.

```python
students = pd.DataFrame(data)
print(students)
print(students.isna())
```

### 10. Count Missing Values

Using the DataFrame from Question 9, calculate the number of missing values in each column and display the total number of missing values in the complete DataFrame.

```python

print(students.isna().sum())

print("Total Missing Values:", students.isna().sum().sum())
```

### 11. Detect Missing Values using isnull()

Create a DataFrame containing missing values in different columns. Use `isnull()` and `notnull()` to display the missing and non-missing values. Compare the results of `isnull()` and `isna()`.

```python
students = pd.DataFrame(data)

print("Missing Values:")
print(students.isnull())

print("\nNon-Missing Values:")
print(students.notnull())

print("\nUsing isna():")
print(students.isna())

print("\nComparison:")
print(students.isnull().equals(students.isna()))
```

### 12. Remove Missing Values

Using a DataFrame containing missing values, demonstrate `dropna()` to remove rows containing at least one missing value. Also demonstrate how to remove columns containing missing values.

```python
students = pd.DataFrame(data)

print("Original DataFrame:")
print(students)


rows_removed = students.dropna()

print("\nAfter Removing Rows:")
print(rows_removed)


columns_removed = students.dropna(axis=1)

print(columns_removed)
```

### 13. Fill Missing Values with a Constant

Create a DataFrame containing missing values in the `Marks` and `City` columns. Replace missing values using suitable constant values with `fillna()`.

```python
students = pd.DataFrame(data)

print("Before Filling:")
print(students)


students["Marks"].fillna(0)
students["City"].fillna("Unknown")

print("\nAfter Filling:")
print(students)
```

### 14. Fill Missing Values using Mean and Median

Create a DataFrame containing missing numerical values. Fill missing values in the `Marks` column using the mean and demonstrate filling another numerical column using the median.

```python
students = pd.DataFrame(data)

# Fill Marks using Mean
mean_marks = students["Marks"].mean()
students["Marks"] = students["Marks"].fillna(mean_marks)

# Fill Semester using Median
median_semester = students["Semester"].median()
students["Semester"] = students["Semester"].fillna(median_semester)

print(students)
```

### 15. Fill Missing Categorical Values using Mode

Create a DataFrame containing missing values in a categorical column such as `City` or `Department`. Fill the missing values using the mode of that column.

```python
students = pd.DataFrame(data)

city_mode = students["City"].mode()[0]
students["City"] = students["City"].fillna(city_mode)
print(students)
```

### 16. Forward Fill and Backward Fill

Create a time-series-like DataFrame containing missing values. Demonstrate forward filling using `ffill()` and backward filling using `bfill()`. Observe the difference between the two methods.

```python
data_time_series = {
    "Date": ["2026-09-01", "2026-09-02", "2026-09-03", 
             "2026-09-04", "2026-09-05"],
    "Marks": [80, np.nan, 90, np.nan, 85]
}

d = pd.DataFrame(data_time_series)
print(d)
forward = d.ffill()

print("\nAfter Forward Fill:")
print(forward)

backward = d.bfill()
print("\nAfter Backward Fill:")
print(backward)
```

### 17. Groupby - Basic Grouping

Create a DataFrame containing `Department`, `Student`, and `Marks`. Group the data by `Department` and calculate the average marks for each department.

```python

students = pd.DataFrame(data)
result = students.groupby("Department")["Marks"].mean()

print(result)
```

### 18. Groupby with Multiple Aggregations

Using a student DataFrame, group the records by `Department` and calculate the minimum, maximum, mean, and count of `Marks` for each department.

```python
students = pd.DataFrame(data)

result = students.groupby("Department")["Marks"].agg( ["min", "max", "mean", "count"])

print(result)
```

### 19. Groupby with Multiple Columns

Create a DataFrame containing `Department`, `Semester`, `Student`, and `Marks`. Group the data by both `Department` and `Semester` and calculate the average marks.

```python
students = pd.DataFrame(data)

result = students.groupby(
    ["Department", "Semester"]
)["Marks"].mean()

print(result)
```

### 20. Named Aggregation using agg()

Using a student DataFrame, group the data by `Department` and use `agg()` with named aggregation to calculate `Maximum_Marks`, `Minimum_Marks`, and `Average_Marks`.

```python
result = students.groupby("Department").agg(
    Maximum_Marks=("Marks", "max"),
    Minimum_Marks=("Marks", "min"),
    Average_Marks=("Marks", "mean")
)

print(result)
```

### 21. Groupby and Count Records

Create a DataFrame containing student details and departments. Use `groupby()` to count the number of students in each department. Also calculate the number of unique cities for each department.

```python

result = students.groupby("Department").agg(
    Student_Count=("Name", "count"),
    Unique_Cities=("City", "nunique")
)

print(result)
```

### 22. Groupby after Handling Missing Data

Create a DataFrame containing `Department`, `Marks`, and `City` with missing values. Handle the missing marks appropriately and then group the data by department to calculate average marks.

```python

mean_marks = students["Marks"].mean()
students["Marks"] = students["Marks"].fillna(mean_marks)


result = students.groupby("Department")["Marks"].mean()

print(result)
```
