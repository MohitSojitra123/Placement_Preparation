### 1. Fetch MCA Faculty Page and Check Status
####  Send an HTTP GET request to https://darshan.ac.in/faculty-list/mca using requests.get(). 
#### Display the HTTP status code to verify a successful connection (200) and print the length of the page source code.

```python
import requests

url="https://darshan.ac.in/faculty-list/mca";

res=requests.get(url);

print(res.status_code)
```

### 2. Extract Web Page Title.
#### Parse the HTML content using BeautifulSoup with 'html.parser'. 
#### Find and display the content of the title tag to identify the webpage.

```python
from bs4 import BeautifulSoup as bs

soup=bs(res.text,"html.parser")

print(soup.title.text)
```

### 3. Scrape Faculty Names using Heading Tags.
#### Locate all faculty names displayed on the page (typically enclosed in h3 tag or specific heading tags within faculty cards).
#### Loop through the results and display the first 5 faculty names cleanly without leading or trailing spaces

```python
name=soup.select("h2.g-mb-0")

list1=[]

for i in name:
    print(i.text.strip())
    list1.append(i.text.strip())
```

### 4. Extract Faculty Designations.
#### Identify the HTML tag or class used for designations (e.g., Assistant Professor, Associate Professor, Head of Department). 
#### Extract and print the designations of the first 5 faculty members.

```python
name=soup.select("span.g-mb-10")

list2=[]

for i in name:
    print(i.text.strip())
    list2.append(i.text.strip())
```

### 5. Scrape Faculty Qualifications.
#### Find the tags or paragraph elements containing qualification details (such as Ph.D., M.Phil, MCA, M.Tech) for faculty members. 
#### Iterate through them and print the qualification strings.

```python
name=soup.select("span.g-px-20")

list3=[]

for i in name:
    print(i.text.strip())
    list3.append(i.text.strip())    
```

### 6. Extract Profile Page Links using CSS Selectors
#### Use soup.select() to find anchor tags leading to individual faculty detail profiles. 
#### Extract and print both the faculty member's name (link text) and their profile URL (href).

```python
name=soup.select("a.g-brd-bottom")

list4=[]

for i in name:
    print(i['href'])
    list4.append("https://darshan.ac.in/"+i['href'])   
```

### 7. Extract Faculty Profile Image URLs.
#### Locate all (img) tags representing faculty photos inside the card containers. 
#### Loop through them and print the image source URL (src) and the alt text attribute for each faculty member.

```python
name=soup.select("img.img-faculty")

list5=[]

for i in name:
    print(i['src'])
    list5.append(i['src'])    
```

### 8. Filter Faculty by Specific Designation or Qualification.
#### Loop through all scraped faculty designations/qualifications and filter only those holding a specific title (e.g., "Assistant Professor" or containing "Ph.D."). 
#### Store and display the filtered results in a Python list.

```python
for i in list1:
    if i.startswith("Prof"):
        print(i)
```

### 9. Extract Faculty Contact / Email Information.
#### Find all contact links on the faculty cards (such as mailto: email links or phone numbers). 
#### Extract the email addresses or telephone details and display them.

```python
url=list4[10]
url="https://darshan.ac.in/"+url
print(url)

res=requests.get(url)
soup1=bs(res.text,"html.parser")

mon=soup1.select("span.g-font-code")

for i in mon:
    print(i.text.strip())

email=soup1.select("span.g-width-80x")

for i in email:
    print(i.text.strip())
    
```

### 10. Create a Pandas DataFrame of MCA Faculty Directory.
#### Collect all faculty Names, Designations, and Profile Links into separate lists. 
#### Combine them into a structured pandas.DataFrame with columns ["Faculty Name", "Designation", "Profile URL"] and display the first 5 rows.

```python
import pandas as pd

df=pd.DataFrame({'Faculty Name':list1,'Designation':list2,"Profile URL":list4})

print(df)

df.to_csv('output.csv', index=False)
```
