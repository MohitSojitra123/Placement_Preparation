```python
import pandas as pd
import numpy as np
```

### 1. Series in Pandas

Create a Pandas Series containing the marks of 8 students. Display the Series, its index, values, and data type.

```python
s= pd.Series([30,40,50,34,45,56,67,45,34,87])
print(s)
```

### 2. Creating Series from Different Data

Create three Pandas Series: i) from a Python list, ii) from a NumPy array, and iii) from a dictionary. Display all three Series.

```python
s= pd.Series([30,40,50,34,45,56,67,45,34,87])
s1=pd.Series(np.array([10,20,30]))
d={'a':10,'b':20}
s2=pd.Series(d)
print("----------list-------------")
print(s)
print("----------numpy array-------------")
print(s1)
print("----------dictionary-------------")
print(s2)
```

### 3. Custom Index in Series

Create a Pandas Series containing the marks of 5 students with student names as the index. Access the marks of two specific students using their labels.

```python
stu={'ram':20,'ravan':30,'rrr':40,'ddd':50,'rt':60}
marks=pd.Series(stu)
print(marks['ram'])
print(marks['ravan'])
```

### 4. Creating Time Series

Create a date range for 10 consecutive days starting from '2026-09-01' and create a Pandas Series containing daily sales values for those dates. Display the resulting time series.

```python
dates=pd.date_range(start='2026-09-01',periods=9)
sales=pd.Series([40,50,60,60,20,60,56,12,45],index=dates)
print(sales)
```

### 5. Time Series with Frequency

Create a time series for the first 6 months of 2026 using a suitable date frequency. Store monthly revenue values in the Series and display the result.

```python
dates=pd.date_range(start='2026',periods=6,freq='ME')
sales=pd.Series([40,50,60,60,20,60],index=dates)
print(sales)
```

### 6. Creating a DataFrame

Create a DataFrame named `students` containing the following columns: RollNo, Name, Department, Marks, and City. Add records for at least 6 students. Display the DataFrame.

```python
d={'rollno':[101,102,103,104,105,106],
   'Name':['raj','rahul','ram','ravan','rohan','ritik'],
   'Department':['CE','CE','MANAGEMENT','MANAGEMENT','CE','MANAGEMENT'],
   'Marks':[30,40,50,60,70,80],
   'City':['rajkot','raipur','raigadh','rampur','rome','ranchi']}
student=pd.DataFrame(d)
print(student)
```

### 7. DataFrame from Dictionary

Create a DataFrame from a dictionary containing information about 5 products: ProductID, ProductName, Category, Price, and Quantity. Display the DataFrame and its columns.

```python
p={'ProductID':[101,102,103,104,105,106],
   'ProductName':['raj','rahul','ram','ravan','rohan','ritik'],
   'Category':['CE','CE','MANAGEMENT','MANAGEMENT','CE','MANAGEMENT'],
   'Price':[300,400,500,600,700,800],
   'Quantity':[10,20,201,305,604,802]}
product=pd.DataFrame(p)
print(product)
```

### 8. Conditional Selection

Using the `students` DataFrame, display: i) students having Marks greater than 70, ii) students belonging to the 'CSE' department, and iii) students having Marks between 60 and 80.

```python
print("Marks greater than 70:-\n",student[student['Marks']>70])
print("Belong to CE:-\n",student[student['Department']=='CE'])
print("Marks between 60 and 80:-\n",student[(student['Marks']>60) & (student['Marks']<80)])
```

### 9. Multiple Conditions

Using the `students` DataFrame, select students who belong to the 'CSE' department and have Marks greater than or equal to 70. Also display students who belong to either 'CSE' or 'IT'.

```python
print("Marks between 60 and 80:-\n",student[(student['Department']=='CE') & (student['Marks']>=70)])
```

### 10. Conditional Selection on DataFrame Columns

From the `products` DataFrame, display products whose Price is greater than 500 and Quantity is less than 20. Then display only the ProductName and Price columns of the selected records.

```python
print(product[(product['Quantity']<20) & (product['Price']>=200)][['ProductName','Price']])
```

### 15. Integrated Pandas Exercise

Create a student DataFrame containing RollNo, Name, Department, Marks, and City. Set RollNo as the index, select students with Marks >= 75, reset the index, and finally save the resulting DataFrame as both CSV and Excel files.

```python
student_data={'rollno':[1,2,3,4],
              'name':['yatharth','sahil','vimal','akshay'],
              'department':['CE','MECH','CE','MECH'],
              'marks':[12,34,56,78],
              'city':['rajkot','jamnagar','rajkot','dhandhuka']}
student=pd.DataFrame(student_data)
student=student.set_index('rollno')
res=student[student['marks']>=75]
print("marks greater than 75\n",res)
student=student.reset_index()
res.to_csv("top_student.csv")
res.to_excel("top_student.xlsx")
```
