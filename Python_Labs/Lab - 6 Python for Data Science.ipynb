### 14) WAP to create (index, value) pairs for prime numbers.

**Sample Input:**
```
T = (4, 7, 9, 11, 15)
```
**Sample Output:**
```
((1, 7), (3, 11))
```

````python
def prime(l):
    for j in range(2,l):
        if l%j==0:
            return False
    else:
        return True

T = (4, 7, 9, 11, 15)
t=()
for i in range(len(T)):
    b=prime(T[i])
    if b:
        k=tuple([i,T[i]])
        t+=(k,)
t
````

### 16) WAP to compute column-wise sum from a tuple of lists.

**Sample Input:**
```
T = ([1, 2, 3], [4, 5, 6], [7, 8, 9])
```

**Sample Output:**
```
12 15 18

```

````python
T = ([1, 2, 3], [4, 5, 6], [7, 8, 9])
result=()
for i in range(len(T)):
    total=0
    for j in range(len(T[i])):
        total+=T[j][i]
    result+=(total,)
print(result)
````
