## Task 14 — Missing Value Handling

**Why not just drop every row with a missing value?** Because we would lose otherwise perfectly good data
in the other 12 columns of that row. Instead we **impute** — fill in a sensible value — based on the
column type:

- **`Attendance` and `StudyHours` (numerical):** filled with the **median**, not the mean, because these
  columns contain a few outliers (Task 18/19 will confirm this) and the median is not affected by extreme
  values the way the mean is.
- **`Gender` and `ParentEducation` (categorical):** filled with the **mode** (the most frequently occurring
  category), since we cannot compute an "average" of text categories — the most common category is the
  most statistically reasonable guess.

**Functions used:**
- **`.fillna(value)`** replaces every missing (`NaN`) entry in a column with the given value.
- **`.median()`** / **`.mode()[0]`** compute the median / the most frequent value of a column (`mode()`
  can return more than one value if there's a tie, so `[0]` picks the first one).

```python
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].median())
df['StudyHours'] = df['StudyHours'].fillna(df['StudyHours'].median())
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
df['ParentEducation'] = df['ParentEducation'].fillna(df['ParentEducation'].mode()[0])

# Verify
print(df.isnull().sum())
```

## Task 19 — IQR-Based Outlier Detection

The **IQR (Interquartile Range) method** is a formula-based way to detect outliers instead of just
eyeballing a box plot:

1. `Q1` = 25th percentile, `Q3` = 75th percentile.
2. `IQR = Q3 - Q1` (the width of the middle 50% of the data).
3. Any value **below** `Q1 - 1.5*IQR` or **above** `Q3 + 1.5*IQR` is flagged as an outlier.

We apply this to `StudyHours` and `Attendance` — the two columns we already suspected from the box plots.

```python
Q1 = df['StudyHours'].quantile(0.25)
Q3 = df['StudyHours'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_study = df[(df['StudyHours'] < lower_bound) | (df['StudyHours'] > upper_bound)]
print("StudyHours bounds:", round(lower_bound,2), "to", round(upper_bound,2))
print("Number of outliers in StudyHours:", len(outliers_study))
outliers_study[['StudentID','StudyHours']]
```

```python
Q1 = df['Attendance'].quantile(0.25)
Q3 = df['Attendance'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_att = df[(df['Attendance'] < lower_bound) | (df['Attendance'] > upper_bound)]
print("Attendance bounds:", round(lower_bound,2), "to", round(upper_bound,2))
print("Number of outliers in Attendance:", len(outliers_att))
outliers_att[['StudentID','Attendance']]
```

**Decision:** these outliers are **capped (winsorized)** rather than removed, because they still represent
real students (an unusually dedicated student who studies 15 hrs/day, or a student who was mostly absent) —
dropping them would throw away real information. Capping brings extreme values to the boundary limit so
they stop distorting the mean/variance while the student's record is still kept.

```python
def cap_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return series.clip(lower=lower_bound, upper=upper_bound)

df['StudyHours'] = cap_outliers(df['StudyHours'])
df['Attendance'] = cap_outliers(df['Attendance'])

print("Outliers after capping:")
print("StudyHours max:", df['StudyHours'].max())
print("Attendance min:", df['Attendance'].min())
```

## Task 25 — Categorical Encoding

Machine Learning algorithms need **numbers**, not text, so categorical columns must be converted:

- **Label Encoding (`LabelEncoder`)** assigns each category an integer (0, 1, 2, ...). It is appropriate
  for **binary** columns (like `Gender`, `Extracurricular`) or **ordinal** columns where categories have a
  natural rank (like `Grade`: A > B > C > D), because the numeric order doesn't mislead the model.

- **One-Hot Encoding (`pd.get_dummies`)** creates a new 0/1 column for each category. It is appropriate for
  **nominal** columns with no natural order (like `City`), because Label Encoding would wrongly imply that,
  say, "Rajkot=0" is somehow "less than" "Surat=3" — which makes no real-world sense for city names.

**`drop_first=True`** in `get_dummies` drops one of the new columns to avoid redundancy (since if you know
all-but-one category is 0, the last one must be 1 anyway) — this avoids the "dummy variable trap".

```python
le_gender = LabelEncoder()
df['Gender_Encoded'] = le_gender.fit_transform(df['Gender'])

le_extra = LabelEncoder()
df['Extracurricular_Encoded'] = le_extra.fit_transform(df['Extracurricular'])

# Grade has a natural order, so we map it manually instead of relying on alphabetical LabelEncoder order
grade_order = {'D': 0, 'C': 1, 'B': 2, 'A': 3}
df['Grade_Encoded'] = df['Grade'].map(grade_order)

df[['Gender','Gender_Encoded','Extracurricular','Extracurricular_Encoded','Grade','Grade_Encoded']].head()
```

```python
df = pd.get_dummies(df, columns=['City', 'ParentEducation'], drop_first=True)
df.head()
```

**Difference between Label Encoding and One-Hot Encoding (in simple terms):**
Label Encoding turns categories into a **single column of numbers** (0, 1, 2...) — simple, but implies an
order/ranking between categories. One-Hot Encoding turns each category into its **own 0/1 column** — no
false ranking is implied, but it creates more columns. Use Label Encoding for ordinal/binary data, and
One-Hot Encoding for nominal data with no order.

## Task 26 — Feature Scaling

Numerical columns can be on very different scales (`Age` is 18-23, `TotalMarks` is 0-100) — some ML
algorithms (like KNN, SVM, gradient descent-based models) are sensitive to this and can be biased toward
columns with larger numeric ranges. **Feature scaling** puts every numerical column on a comparable scale.

- **`StandardScaler`** (standardisation) transforms a column to have **mean = 0** and **standard
  deviation = 1**. Good when the data roughly follows a normal (bell-curve) distribution.
- **`MinMaxScaler`** (normalisation) squeezes every value into a fixed **0 to 1** range. Good when we want
  a bounded range regardless of distribution shape, but it's more sensitive to outliers.

We scale `Attendance` and `StudyHours` (both continuous numerical features) using both methods, into new
columns, so the original values are preserved for comparison.

```python
scaler_std = StandardScaler()
df[['Attendance_Std','StudyHours_Std']] = scaler_std.fit_transform(df[['Attendance','StudyHours']])

df[['Attendance','Attendance_Std','StudyHours','StudyHours_Std']].head()
```

```python
scaler_mm = MinMaxScaler()
df[['Attendance_MinMax','StudyHours_MinMax']] = scaler_mm.fit_transform(df[['Attendance','StudyHours']])

df[['Attendance','Attendance_MinMax','StudyHours','StudyHours_MinMax']].head()
```

**Difference between Standardisation and Normalisation (in simple terms):**
Standardisation re-centers data around 0 based on the standard deviation (values can go below 0 or above 1);
Normalisation forces all values strictly between 0 and 1 based on the min and max. Standardisation is
preferred when a column is roughly bell-shaped or has some outliers; Normalisation is preferred when you
need bounded values (e.g. for neural networks) and the data has no extreme outliers.

## Viva / Oral Questions — Simple Explanations

Below are short, exam-friendly answers to the standard viva questions for this assignment.

**1. What is Data Science?**
Data Science is the field of extracting useful knowledge and insights from data using statistics,
programming, and domain knowledge — combining math, coding, and business understanding.

**2. What is Exploratory Data Analysis (EDA)?**
EDA is the process of examining a dataset — through summary statistics and visualizations — to understand
its structure, patterns, and problems *before* building any model.

**3. Why is EDA important before Machine Learning?**
Because a model is only as good as the data it's trained on. EDA reveals missing values, outliers, wrong
data types, and relationships between variables — all of which affect model accuracy if ignored.

**4. What is the difference between a row and a column?**
A **row** represents one single record/observation (e.g., one student). A **column** represents one
attribute/feature measured across all records (e.g., everyone's `Age`).

**5. What is the purpose of `head()`, `tail()`, `info()`, and `describe()`?**
`head()`/`tail()` preview the start/end of the data. `info()` shows column types and non-null counts.
`describe()` gives statistical summaries (mean, min, max, etc.) of the columns.

**6. What is a missing value and what are its common causes?**
A missing value is an empty/unrecorded entry in the data (shown as `NaN`). Common causes: data entry
errors, non-response in surveys, sensor failure, or merging datasets that don't perfectly align.

**7. What is the difference between `dropna()` and `fillna()`?**
`dropna()` **removes** rows/columns containing missing values. `fillna()` **fills in** missing values with
a chosen value (mean, median, mode, or a constant) instead of deleting data.

**8. When is median preferable to mean for imputation?**
When the column has outliers or a skewed distribution — the mean gets pulled towards extreme values, while
the median stays close to the "typical" value.

**9. Why is mode useful for categorical variables?**
Because categorical data (like "Male"/"Female") has no numeric average — the mode (most frequent category)
is the most statistically sensible way to fill a gap.

**10. What is univariate analysis?**
Studying **one variable at a time** — its distribution, spread, and common values — without looking at its
relationship with other variables.

**11. What is bivariate analysis?**
Studying the **relationship between two variables** at a time, e.g. how `StudyHours` relates to
`TotalMarks`.

**12. What is the purpose of a histogram?**
A histogram shows the **distribution/shape** of a numerical variable — where most values cluster, whether
it's skewed, and whether it's roughly symmetric (bell-shaped) or not.

**13. What does a box plot show?**
A box plot shows the median, the middle 50% of data (the box), the overall spread (the whiskers), and any
outliers (individual points beyond the whiskers) — all in one compact picture.

**14. What are outliers?**
Outliers are data points that are unusually far from the rest of the data — either much higher or much
lower than the typical range.

**15. What is IQR and how is it used to detect outliers?**
IQR (Interquartile Range) = Q3 - Q1, the width of the middle 50% of data. Any value below `Q1 - 1.5*IQR` or
above `Q3 + 1.5*IQR` is flagged as a statistical outlier.

**16. What is a scatter plot?**
A scatter plot places one dot per record based on two numerical values, letting us visually see whether the
two variables move together (correlated) or not.

**17. What is correlation?**
Correlation measures how strongly two numerical variables move together — whether an increase in one tends
to come with an increase (or decrease) in the other.

**18. What does a correlation value of +1 or -1 indicate?**
+1 means a perfect positive relationship (both increase together); -1 means a perfect negative relationship
(one increases as the other decreases). 0 means no linear relationship at all.

**19. What is categorical encoding?**
The process of converting text/category values into numbers so that Machine Learning algorithms — which
only understand numbers — can use them.

**20. What is the difference between Label Encoding and One-Hot Encoding?**
Label Encoding assigns each category a single number (0,1,2...) — good for ordered/binary categories.
One-Hot Encoding creates a separate 0/1 column per category — good for unordered (nominal) categories,
since it avoids implying a false ranking.

**21. What is feature scaling?**
The process of bringing numerical columns onto a similar scale/range, so that no single column dominates a
model just because its raw numbers happen to be larger.

**22. What is the difference between standardisation and normalisation?**
Standardisation rescales data to have mean 0 and standard deviation 1 (values can be negative). Normalisation
(Min-Max) rescales data strictly into the 0-to-1 range.

**23. Why can duplicate records affect analysis?**
Duplicates artificially inflate counts and can skew averages/statistics towards whatever values are
repeated, making the dataset look different from reality.

**24. Why should preprocessing decisions be based on data characteristics?**
Because a "one-size-fits-all" approach can damage the data — e.g. using mean imputation on a skewed column,
or dropping rows that actually contain valid outliers, can introduce bias and reduce the dataset's quality
and reliability for later analysis or modeling.
