### 5. WAP to check whether the given year is Leap year or not. 
#### [If a year can be divisible by 4 but not divisible by 100 then it is leap year but if it is divisible by 400 then it is leap year].

```python
year=int(input("Enter Any Year To Check Year Is Leap Or Not : "))

if (year % 400 == 0 ) or (year%4 ==0 and year %100 != 0):
    print(f"{year} is Leap : ")
else: 
    print (f"{year} is Not Leap : ")
    
```

### 7. WAP to implement simple Calculator which performs (add,sub,mul,div) of two numbers based on user input.

```python
a=int(input("Enter Number 1: "))
b=int(input("Enter Number 2 : "))

operation=input("Enter + , - , * , / : ")

match operation:
    case "+":
        print(f"{a}+{b} = {a+b}")
    case "-":
        print(f"{a} - {b} = {a-b}")
    case "*":
        print(f"{a} * {b} = {a*b} ")
    case "/":
        print(f"{a} / {b} = {a/b}")
    case _:
         print("Invalid Input : ")
        
```

### 8. WAP to calculate electricity bill based on following criteria. Which takes the unit from the user.
#### 	  -First 1 to 50 units – Rs. 2.60/unit
#### 	  -Next 50 to 100 units – Rs. 3.25/unit
#### 	  -Next 100 to 200 units – Rs. 5.26/unit
####	  -200 units – Rs. 8.45/unit

```python
unit=float(input("Enter Total Unit : "))

if(unit<=50):
    print(f"Unit -> {unit*2.60}")
elif (unit>50 and unit<=100):
    print(f"Unit -> {(50*2.60) + ((unit-50)*3.25)}")
elif (unit>100 and unit<=200):
    print(f"Unit -> {(50*2.60)+(50*3.25)+((unit-100)*5.26)}")
else:
    print(f"Unit -> {(50*2.60)+(50*3.25)+(100*5.26)+((unit-200)*8.45)}")
```
