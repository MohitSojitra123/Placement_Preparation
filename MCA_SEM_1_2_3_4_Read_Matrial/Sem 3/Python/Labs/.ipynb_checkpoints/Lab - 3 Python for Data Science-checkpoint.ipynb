### 3.WAP to print odd numbers between 1 to n.

```python
n=int(input("Enter N : "))

print("Odd Number : ")
for i in range(1,n+1):
    if (i%2!=0):
        print(i)
```

### 4. WAP to print numbers between two given numbers which is divisible by 2 but not divisible by 3.

```python
start=int(input("Enter Starting Number : "))
end=int(input("Enter Ending Number : "))

print("Divisible By 2 Not Divisible By 3 :")
for i in range(start,end):
    if (i%2==0 and i%3!=0):
        print(i)
```

### 7. WAP to print sum of series 1 – 2 + 3 – 4 + 5 – 6 + 7 ... n.

```python
n=int(input("Enter N : "))

total=0

for i in range(1,n+1,2):
    print(f"{i}-{i+1}",end=" + ")
    total+=(i-(i+1))

print(" ==> ",total)
    
```

### 9. WAP to find factorial of the given number.

```python
n=int(input("Enter N : "))

fact=1

for i in range(1,n+1):
    fact*=i

print(fact)    
```

### 11. WAP to find whether the given number is prime or not.

```python
n=int(input("Enter Number To Check Number Is Prime Or Not : "))

isprime=True

for i in range(2,n):
    if(n%i==0):
        isprime=False
        break

if(isprime):
    print("Prime Number :")
else:
    print("Not Prime Number : ")
```

### 12. WAP to print sum of digits of given number.

```python
n=int(input("Enter N : "))

total=0

while(n!=0):
    total+=n%10 
    n//=10

print(total)
```

### 13. WAP to check whether the given number is palindrome or not.

```python
n=int(input("Enter Number To Check Number Is Palindrom Or Not : "))

n2=n;
reverse_num=0;

while(n2!=0):
    reverse_num=((reverse_num*10)+(n2%10))
    n2//=10

if n==reverse_num:
    print("Number Is Palindrom : ")
else:
    print("Number Is Not Palindrom : ")
```
