### Q2. What is the difference between interpreted and compiled languages?

**Answer:** A **compiled** language (C, C++) is translated entirely to machine code before running; it runs fast, but errors are found at compile time. An **interpreted** language (Python, JS) is executed line by line by an interpreter. Python actually compiles source to bytecode (`.pyc`) first and the Python Virtual Machine interprets that bytecode.

### Q39. Explain operator precedence in Python.

### Q40. What is the difference between x += 1 and x = x + 1?

```python
a = [1]; b = a
a += [2];  print(b)       # b changed (in-place)
a = [1]; b = a
a = a + [2]; print(b)     # b unchanged
```

### Q44. Explain short-circuit evaluation with an example.

### Q60. Write a program to calculate an electricity bill using different price slabs.

**Answer:** Sample slabs: first 100 units ₹3, next 100 units ₹5, next 100 units ₹7, above 300 units ₹10.

```python
def bill(units):
    slabs = [(100, 3), (100, 5), (100, 7)]
    total, remaining = 0, units
    for size, rate in slabs:
        used = min(remaining, size)
        total += used * rate
        remaining -= used
    return total + remaining * 10
for u in (50, 150, 250, 400):
    print(u, "units -> Rs", bill(u))
```

### Q161. Write a program to find the longest consecutive increasing run in a list.

```python
def longest_run(l):
    if not l: return []
    best_s = cur_s = 0; best_len = cur_len = 1
    for i in range(1, len(l)):
        if l[i] > l[i-1]:
            cur_len += 1
        else:
            cur_s, cur_len = i, 1
        if cur_len > best_len:
            best_s, best_len = cur_s, cur_len
    return l[best_s:best_s + best_len]
print(longest_run([1, 3, 5, 4, 7, 8, 9, 2]))
```

### Q162. Write a program to find the maximum sum of a contiguous sublist.

**Answer:** Kadane's algorithm, O(n).

```python
def max_subarray(l):
    best = cur = l[0]
    for x in l[1:]:
        cur = max(x, cur + x)
        best = max(best, cur)
    return best
print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))
```

### Q165. Write a program to rearrange a list so that positive and negative numbers alternate when possible.

```python
def alternate(l):
    pos = [x for x in l if x >= 0]
    neg = [x for x in l if x < 0]
    out, i, j, take_pos = [], 0, 0, True
    while i < len(pos) and j < len(neg):
        if take_pos: out.append(pos[i]); i += 1
        else: out.append(neg[j]); j += 1
        take_pos = not take_pos
    return out + pos[i:] + neg[j:]
print(alternate([1, 2, 3, -4, -1, 4]))
```

### Q264. Write a recursive function to calculate factorial.

### Q265. Write a recursive function to calculate the nth Fibonacci number.

### Q270. Write a recursive function to calculate a number raised to an integer power.

```python
def power(x, n):
    if n == 0: return 1
    if n < 0: return 1 / power(x, -n)
    half = power(x, n // 2)
    return half * half if n % 2 == 0 else half * half * x
print(power(2, 10), power(2, -2), power(3, 5))
```

### Q272. Explain the difference between recursion and iteration.

**Answer:** Iteration repeats using loops with explicit state; recursion repeats through self-calls and keeps state on the call stack. ### Q273. What is the recursion limit in Python, and why does it exist?

### Q274. Trace the call stack for a recursive factorial function.

**Answer:** For `fact(3)`: calls go down `fact(3) → fact(2) → fact(1)` (base case returns 1); then values return upward: `fact(2)=2*1=2`, `fact(3)=3*2=6`.

```python
def fact(n, depth=0):
    print("  " * depth + f"fact({n}) called")
    r = 1 if n <= 1 else n * fact(n - 1, depth + 1)
    print("  " * depth + f"fact({n}) returns {r}")
    return r
fact(3)
```

### Q338. Write a program to generate all prime numbers in a given range.

**Answer:** Sieve of Eratosthenes.

```python
def primes_in_range(lo, hi):
    sieve = [True] * (hi + 1); sieve[:2] = [False, False]
    for i in range(2, int(hi ** 0.5) + 1):
        if sieve[i]:
            for j in range(i * i, hi + 1, i): sieve[j] = False
    return [n for n in range(max(lo, 2), hi + 1) if sieve[n]]
print(primes_in_range(10, 60))
```

### Q340. Write a program to convert a decimal integer to binary without using bin().

```python
def to_binary(n):
    if n == 0: return "0"
    sign, n, bits = ("-" if n < 0 else ""), abs(n), []
    while n:
        bits.append(str(n % 2)); n //= 2
    return sign + "".join(reversed(bits))
print(to_binary(10), to_binary(255), to_binary(0), to_binary(-5))
```

### Q390. Explain the diamond inheritance problem and how Python's MRO addresses it.

**Answer:** In a diamond (D inherits B and C, both inheriting A) it is ambiguous which version of an inherited method to call, and A's initialiser might run twice. Python's C3 MRO linearises the classes so each appears once and consistently; with cooperative `super()` every `__init__` runs exactly once.

```python
class A:
    def __init__(self): print("A")
class B(A):
    def __init__(self): print("B"); super().__init__()
class C(A):
    def __init__(self): print("C"); super().__init__()
class D(B, C):
    def __init__(self): print("D"); super().__init__()
D()
print([k.__name__ for k in D.mro()])
```

### Q401. What are magic methods, also called dunder methods?

### Q402. What is the difference between __str__() and __repr__()?

**Answer:** `__str__` → readable output for end users (`print`, `str()`); `__repr__` → unambiguous developer representation (`repr()`, REPL); `str()` falls back to `__repr__` if `__str__` is missing.

```python
class P:
    def __init__(self, x): self.x = x
    def __str__(self): return f"Point at {self.x}"
    def __repr__(self): return f"P({self.x!r})"
p = P(3); print(p, repr(p), [p])
```

**Answer:** It defines the result of `len(obj)` and also gives truthiness (0 → falsy).

```python
class Bag:
    def __init__(self, *i): self.i = list(i)
    def __len__(self): return len(self.i)
print(len(Bag(1, 2, 3)), bool(Bag()))
```

### Q404. How do __eq__() and __ne__() affect comparisons?

**Answer:** `__eq__` defines `==`; `__ne__` defines `!=` (Python 3 derives it from `__eq__` by default). Defining `__eq__` without `__hash__` makes the class unhashable.

```python
class P:
    def __init__(self, x): self.x = x
    def __eq__(self, o): return isinstance(o, P) and self.x == o.x
    def __hash__(self): return hash(self.x)
print(P(1) == P(1), P(1) != P(2), len({P(1), P(1)}))
```

### Q405. What are __lt__(), __le__(), __gt__(), and __ge__() used for?

**Answer:** They implement `<`, `<=`, `>`, `>=`, enabling sorting and comparison of objects. `functools.total_ordering` can generate the rest from `__eq__` plus one.

```python
from functools import total_ordering
@total_ordering
class V:
    def __init__(self, n): self.n = n
    def __eq__(self, o): return self.n == o.n
    def __lt__(self, o): return self.n < o.n
    def __repr__(self): return f"V({self.n})"
print(V(1) < V(2), V(3) >= V(2), sorted([V(3), V(1), V(2)]))
```

## Topic 31: Decorators (416–425)

### Q416. What is a decorator in Python?

**Answer:** A decorator is a callable that takes a function (or class), wraps it with extra behaviour and returns the replacement. `@deco` above `def f` is shorthand for `f = deco(f)`.

### Q418. What is a closure, and how does it help implement decorators?

**Answer:** A closure is an inner function that remembers variables of its enclosing scope after the outer function has returned. A decorator's `wrapper` is a closure that remembers the wrapped function `fn`.

```python
def deco(fn):
    def wrapper(*a, **k):
        print("calling", fn.__name__)
        return fn(*a, **k)
    return wrapper
@deco
def add(a, b): return a + b
print(add(1, 2), add.__closure__[0].cell_contents.__name__)
```

### Q419. What is the purpose of functools.wraps in a decorator?

**Answer:** `@wraps(fn)` copies the wrapped function's metadata (`__name__`, `__doc__`, `__wrapped__`, ...) onto the wrapper so introspection/help/debugging still show the original function.

```python
from functools import wraps
def deco(fn):
    @wraps(fn)
    def w(*a, **k): return fn(*a, **k)
    return w
@deco
def f():
    "Docstring of f"
print(f.__name__, f.__doc__)
```

### Q420. Write a decorator that prints a message before and after a function executes.

### Q421. Write a decorator that measures a function's execution time using the standard library.

```python
import time
from functools import wraps
def timer(fn):
    @wraps(fn)
    def wrapper(*a, **k):
        start = time.perf_counter()
        try: return fn(*a, **k)
        finally: print(f"{fn.__name__} took {time.perf_counter() - start:.6f}s")
    return wrapper
@timer
def work(n): return sum(range(n))
print(work(1_000_000))
```

### Q422. Write a decorator that validates the types of function arguments.

```python
from functools import wraps
def check_types(*types):
    def deco(fn):
        @wraps(fn)
        def wrapper(*args):
            for a, t in zip(args, types):
                if not isinstance(a, t):
                    raise TypeError(f"{a!r} is not {t.__name__}")
            return fn(*args)
        return wrapper
    return deco
@check_types(int, int)
def add(a, b): return a + b
print(add(1, 2))
try: add(1, "2")
except TypeError as e: print("TypeError:", e)
```

### Q423. Write a decorator that counts how many times a function is called.

```python
from functools import wraps
def count_calls(fn):
    @wraps(fn)
    def wrapper(*a, **k):
        wrapper.calls += 1
        return fn(*a, **k)
    wrapper.calls = 0
    return wrapper
@count_calls
def hi(): pass
hi(); hi(); hi()
print(hi.calls)
```

### Q424. Write a decorator that caches results for repeated calls.

```python
from functools import wraps
def memoize(fn):
    cache = {}
    @wraps(fn)
    def wrapper(*args):
        if args not in cache:
            cache[args] = fn(*args)
        return cache[args]
    wrapper.cache = cache
    return wrapper
@memoize
def fib(n): return n if n < 2 else fib(n - 1) + fib(n - 2)
print(fib(60), len(fib.cache))
# standard library equivalent: functools.lru_cache / functools.cache
```

### Q425. Write a decorator that accepts its own arguments and uses them to control a function's behavior.

**Answer:** A decorator factory: three nested levels (factory → decorator → wrapper).

```python
from functools import wraps
def repeat(times):
    def deco(fn):
        @wraps(fn)
        def wrapper(*a, **k):
            result = None
            for _ in range(times): result = fn(*a, **k)
            return result
        return wrapper
    return deco
@repeat(3)
def hello(): print("Hello!")
hello()
```

### Q428. Explain the difference between global and nonlocal in nested functions.

### Q429. Predict the behavior of a nested function that reads a variable from its enclosing function.

**Answer:** It can read it freely (via LEGB) without any keyword. But if it *assigns* to the name, the name becomes local and reading earlier raises `UnboundLocalError`.

### Q445. Explain the limitations of deeply recursive algorithms and propose an iterative alternative.

**Answer:** Python limits recursion depth (~1000), each frame costs memory/time, and there is no tail-call optimisation, so deep recursion risks `RecursionError`. Fix: use loops or an explicit stack.

```python
def depth_sum_recursive(n): return 0 if n == 0 else n + depth_sum_recursive(n - 1)
def depth_sum_iter(n):
    total = 0
    for i in range(1, n + 1): total += i
    return total
print(depth_sum_iter(100000))
try: depth_sum_recursive(100000)
except RecursionError: print("recursive version fails")
# explicit stack for tree/nested-list traversal
def flatten_iter(data):
    stack, out = [iter(data)], []
    while stack:
        for x in stack[-1]:
            if isinstance(x, list): stack.append(iter(x)); break
            out.append(x)
        else: stack.pop()
    return out
print(flatten_iter([1, [2, [3, [4]]], 5]))
```

### Q476. Implement a custom iterable class that produces square numbers.

```python
class Squares:
    def __init__(self, n): self.n = n
    def __iter__(self):
        for i in range(1, self.n + 1): yield i * i
s = Squares(5)
print(list(s), list(s))
```

### Q477. Write a generator that yields prime numbers indefinitely.

```python
def primes():
    found = []
    n = 2
    while True:
        if all(n % p for p in found if p * p <= n):
            found.append(n); yield n
        n += 1
from itertools import islice
print(list(islice(primes(), 12)))
```

### Q478. Write a generator that reads a sequence in batches of a given size.

```python
def batches(seq, size):
    batch = []
    for x in seq:
        batch.append(x)
        if len(batch) == size: yield batch; batch = []
    if batch: yield batch
print(list(batches(range(10), 4)))
```

### Q479. Explain the difference between generator exhaustion and generator reset.

**Answer:** A generator can only run forward; once exhausted it can't be rewound ("reset"). To iterate again you must call the generator function again (or use an iterable class whose `__iter__` returns a new iterator, or `itertools.tee`).

### Q482. Explain lazy evaluation and compare it with eager evaluation.

**Answer:** Lazy evaluation computes values only when needed (generators, `range`, `map`, `filter`, `zip`) — low memory, works for infinite streams, can short-circuit. Eager evaluation computes everything immediately (lists) — allows indexing/reuse but uses memory up front.

```python
import itertools
lazy = (x * x for x in itertools.count(1))     # infinite, no computation yet
print(next(lazy), next(lazy), next(lazy))
```

### Q484. Implement a custom iterator that supports reverse traversal through a sequence.

```python
class Traversal:
    def __init__(self, seq): self.seq = seq
    def __iter__(self): return iter(self.seq)
    def __reversed__(self):
        i = len(self.seq) - 1
        while i >= 0:
            yield self.seq[i]; i -= 1
t = Traversal([1, 2, 3, 4])
print(list(t), list(reversed(t)))
```

### Q485. Explain the advantages and limitations of generators when processing large files.

**Answer:** **Advantages**: constant memory (one line at a time), can start producing results immediately, composes into pipelines. **Limitations**: single pass (no random access/len), can't rewind, harder to debug, and the file must stay open while iterating (consume inside the `with` or via a generator that owns the file).

```python
def long_lines(path, min_len):
    with open(path) as f:
        for line in f:
            if len(line) >= min_len: yield line.rstrip()
print(list(long_lines("sample.txt", 20)))
```

### Q488. What are __getattr__() and __getattribute__()?

### Q489. What is the purpose of __slots__, and what trade-offs does it introduce?

**Answer:** `__slots__` declares a fixed set of attributes, removing the per-instance `__dict__` — lower memory and slightly faster access. Trade-offs: no dynamic new attributes, no `__weakref__` by default, inheritance subtleties, and incompatibility with some features.

### Q490. What is a descriptor in Python?

**Answer:** An object (defined in a class) that implements `__get__`, `__set__` and/or `__delete__` and thereby customises attribute access. Properties, methods, `classmethod`, `staticmethod` and `__slots__` are implemented with descriptors.

```python
class Positive:
    def __set_name__(self, owner, name): self.name = "_" + name
    def __get__(self, obj, objtype=None): return getattr(obj, self.name)
    def __set__(self, obj, v):
        if v <= 0: raise ValueError("must be positive")
        setattr(obj, self.name, v)
class Item:
    price = Positive()
    def __init__(self, p): self.price = p
i = Item(5); print(i.price)
try: i.price = -1
except ValueError as e: print(e)
```

### Q491. Explain the descriptor methods __get__(), __set__(), and __delete__().

**Answer:** `__get__(self, obj, objtype)` runs when the attribute is read; `__set__(self, obj, value)` when assigned; `__delete__(self, obj)` on `del`. An object with `__set__` or `__delete__` is a *data descriptor* (takes precedence over instance `__dict__`); only `__get__` makes it a *non-data descriptor*.

```python
class Logged:
    def __get__(self, obj, t=None): print("get"); return 42
    def __set__(self, obj, v): print("set", v)
    def __delete__(self, obj): print("delete")
class C: attr = Logged()
c = C(); c.attr; c.attr = 7; del c.attr
```

### Q492. What is a metaclass, and what is its relationship to the type class?

**Answer:** A metaclass is the class of a class — it controls how classes are created. By default every class is an instance of `type`; custom metaclasses subclass `type` and are set via `metaclass=...`.

```python
class Meta(type):
    def __new__(mcs, name, bases, ns):
        ns["created_by"] = "Meta"
        return super().__new__(mcs, name, bases, ns)
class A(metaclass=Meta): pass
print(A.created_by, type(A), type(type))
```

### Q493. What is the difference between __new__() and __init__()?

**Answer:** `__new__(cls, ...)` is a static method that **creates and returns** the instance (called first); `__init__(self, ...)` **initialises** the already-created instance (returns None). `__new__` is needed for immutable types and singletons.

```python
class Singleton:
    _inst = None
    def __new__(cls):
        if cls._inst is None: cls._inst = super().__new__(cls)
        return cls._inst
print(Singleton() is Singleton())
class Upper(str):
    def __new__(cls, v): return super().__new__(cls, v.upper())
print(Upper("abc"))
```

### Q494. What are dataclasses, and when might @dataclass be useful?

### Q496. Library Management System: Design classes for books, members, and loans. Implement borrowing, returning, availability checks, and validation using OOP.

```python
from datetime import date, timedelta
class LibraryError(Exception): pass
class Book:
    def __init__(self, isbn, title, author, copies=1):
        if copies < 0: raise ValueError("copies cannot be negative")
        self.isbn, self.title, self.author = isbn, title, author
        self.total, self.available = copies, copies
    def __repr__(self): return f"Book({self.title!r}, available={self.available}/{self.total})"
class Member:
    MAX_LOANS = 3
    def __init__(self, member_id, name): self.member_id, self.name, self.loans = member_id, name, []
class Loan:
    def __init__(self, book, member, days=14):
        self.book, self.member = book, member
        self.issued = date.today(); self.due = self.issued + timedelta(days=days)
        self.returned = None
class Library:
    def __init__(self): self.books, self.members, self.loans = {}, {}, []
    def add_book(self, book): self.books[book.isbn] = book
    def add_member(self, m): self.members[m.member_id] = m
    def is_available(self, isbn): return isbn in self.books and self.books[isbn].available > 0
    def borrow(self, member_id, isbn):
        if member_id not in self.members: raise LibraryError("Unknown member")
        if isbn not in self.books: raise LibraryError("Unknown book")
        m, b = self.members[member_id], self.books[isbn]
        if len(m.loans) >= Member.MAX_LOANS: raise LibraryError("Loan limit reached")
        if any(l.book is b for l in m.loans): raise LibraryError("Already borrowed this book")
        if not self.is_available(isbn): raise LibraryError("No copies available")
        loan = Loan(b, m); b.available -= 1; m.loans.append(loan); self.loans.append(loan)
        return loan
    def return_book(self, member_id, isbn):
        m = self.members[member_id]
        for loan in m.loans:
            if loan.book.isbn == isbn:
                loan.returned = date.today(); loan.book.available += 1; m.loans.remove(loan)
                return loan
        raise LibraryError("This member has not borrowed that book")

lib = Library()
lib.add_book(Book("111", "Python Basics", "A. Author", copies=1)); lib.add_book(Book("222", "Algorithms", "B. Writer", 2))
lib.add_member(Member(1, "Asha")); lib.add_member(Member(2, "Ravi"))
lib.borrow(1, "111"); print(lib.books["111"])
try: lib.borrow(2, "111")
except LibraryError as e: print("Error:", e)
try: lib.borrow(1, "111")
except LibraryError as e: print("Error:", e)
lib.return_book(1, "111"); print(lib.books["111"], lib.is_available("111"))
```

### Q497. Banking System: Build an account hierarchy with savings and current accounts, transaction history, deposit, withdrawal, transfer, and custom exceptions.

```python
class BankError(Exception): pass
class InsufficientFundsError(BankError): pass
class InvalidAmountError(BankError): pass
class Account:
    _next = 1000
    def __init__(self, owner, balance=0):
        self.owner, self._balance = owner, 0
        self.number = Account._next; Account._next += 1
        self.history = []
        if balance: self.deposit(balance)
    @property
    def balance(self): return self._balance
    def _check(self, amt):
        if not isinstance(amt, (int, float)) or amt <= 0: raise InvalidAmountError("Amount must be positive")
    def _limit(self): return 0                      # overdraft allowance
    def deposit(self, amt):
        self._check(amt); self._balance += amt; self.history.append(("deposit", amt, self._balance))
    def withdraw(self, amt):
        self._check(amt)
        if amt > self._balance + self._limit(): raise InsufficientFundsError(f"Cannot withdraw {amt}; balance {self._balance}")
        self._balance -= amt; self.history.append(("withdraw", amt, self._balance))
    def transfer(self, other, amt):
        self.withdraw(amt); other.deposit(amt)
        self.history.append(("transfer out", amt, self._balance))
    def __repr__(self): return f"{type(self).__name__}(#{self.number}, {self.owner}, {self._balance})"
class SavingsAccount(Account):
    def __init__(self, owner, balance=0, rate=0.04): super().__init__(owner, balance); self.rate = rate
    def add_interest(self): self.deposit(round(self._balance * self.rate, 2))
class CurrentAccount(Account):
    def __init__(self, owner, balance=0, overdraft=500): super().__init__(owner, balance); self.overdraft = overdraft
    def _limit(self): return self.overdraft

s = SavingsAccount("Asha", 1000); c = CurrentAccount("Asha", 200)
s.add_interest(); c.withdraw(600); s.transfer(c, 300)
print(s, c)
try: s.withdraw(10**6)
except InsufficientFundsError as e: print("Error:", e)
try: s.deposit(-5)
except InvalidAmountError as e: print("Error:", e)
for h in s.history: print(h)
```

### Q499. Custom Data Structures: Implement a stack, queue, and singly linked list using classes, including insertion, deletion, searching, traversal, and edge-case handling.

```python
class Stack:
    def __init__(self): self._d = []
    def push(self, x): self._d.append(x)
    def pop(self):
        if not self._d: raise IndexError("pop from empty stack")
        return self._d.pop()
    def peek(self):
        if not self._d: raise IndexError("peek from empty stack")
        return self._d[-1]
    def is_empty(self): return not self._d
    def __len__(self): return len(self._d)

class _QNode:
    def __init__(self, v): self.v, self.next = v, None
class Queue:
    def __init__(self): self.head = self.tail = None; self._n = 0
    def enqueue(self, x):
        n = _QNode(x)
        if self.tail: self.tail.next = n
        else: self.head = n
        self.tail = n; self._n += 1
    def dequeue(self):
        if not self.head: raise IndexError("dequeue from empty queue")
        v = self.head.v; self.head = self.head.next
        if not self.head: self.tail = None
        self._n -= 1; return v
    def __len__(self): return self._n

class Node:
    def __init__(self, data, next=None): self.data, self.next = data, next
class LinkedList:
    def __init__(self): self.head = None
    def insert_front(self, x): self.head = Node(x, self.head)
    def insert_end(self, x):
        n = Node(x)
        if not self.head: self.head = n; return
        cur = self.head
        while cur.next: cur = cur.next
        cur.next = n
    def delete(self, x):
        if not self.head: return False
        if self.head.data == x: self.head = self.head.next; return True
        cur = self.head
        while cur.next and cur.next.data != x: cur = cur.next
        if cur.next: cur.next = cur.next.next; return True
        return False
    def search(self, x):
        cur, i = self.head, 0
        while cur:
            if cur.data == x: return i
            cur, i = cur.next, i + 1
        return -1
    def __iter__(self):
        cur = self.head
        while cur: yield cur.data; cur = cur.next
    def __repr__(self): return " -> ".join(map(str, self)) + " -> None"

s = Stack(); s.push(1); s.push(2); print(s.pop(), s.peek(), len(s))
try: Stack().pop()
except IndexError as e: print("Error:", e)
q = Queue(); q.enqueue("a"); q.enqueue("b"); print(q.dequeue(), q.dequeue(), len(q))
try: q.dequeue()
except IndexError as e: print("Error:", e)
ll = LinkedList(); ll.insert_end(2); ll.insert_end(3); ll.insert_front(1)
print(ll, ll.search(3), ll.search(9), ll.delete(2), ll.delete(99), ll, list(ll))
```

### Q500. Mini Python Project — Expense Tracker: Build a command-line expense tracker using only Core Python. Support adding, editing, deleting, searching, categorizing, and summarizing expenses; save and load data from a file; validate input; and use OOP, comprehensions, and exception handling.

**Answer:** The class-based core below is fully runnable in Jupyter. The last cell contains the interactive command-line menu (`input()`); a scripted demo runs first.

```python
import json, os
from datetime import datetime
class ExpenseError(Exception): pass
class Expense:
    def __init__(self, id, title, amount, category, date=None):
        if not str(title).strip(): raise ExpenseError("Title cannot be empty")
        try: amount = float(amount)
        except (TypeError, ValueError): raise ExpenseError("Amount must be a number")
        if amount <= 0: raise ExpenseError("Amount must be positive")
        date = date or datetime.now().strftime("%Y-%m-%d")
        try: datetime.strptime(date, "%Y-%m-%d")
        except ValueError: raise ExpenseError("Date must be YYYY-MM-DD")
        self.id, self.title, self.amount = id, str(title).strip(), amount
        self.category, self.date = str(category).strip().title() or "General", date
    def to_dict(self): return dict(id=self.id, title=self.title, amount=self.amount, category=self.category, date=self.date)
    def __str__(self): return f"#{self.id:<3}{self.date}  {self.category:<12}{self.title:<20}{self.amount:>10.2f}"

class ExpenseTracker:
    def __init__(self, path="expenses.json"):
        self.path, self.expenses, self._next_id = path, {}, 1
        self.load()
    def add(self, title, amount, category="General", date=None):
        e = Expense(self._next_id, title, amount, category, date)
        self.expenses[e.id] = e; self._next_id += 1; self.save(); return e
    def edit(self, eid, **changes):
        if eid not in self.expenses: raise ExpenseError(f"No expense with id {eid}")
        d = self.expenses[eid].to_dict(); d.update({k: v for k, v in changes.items() if v is not None})
        self.expenses[eid] = Expense(**d); self.save()
    def delete(self, eid):
        if self.expenses.pop(eid, None) is None: raise ExpenseError(f"No expense with id {eid}")
        self.save()
    def search(self, keyword="", category=None):
        k = keyword.lower()
        return [e for e in self.expenses.values()
                if k in e.title.lower() and (category is None or e.category.lower() == category.lower())]
    def summary(self):
        by_cat = {}
        for e in self.expenses.values(): by_cat[e.category] = by_cat.get(e.category, 0) + e.amount
        return {"total": sum(e.amount for e in self.expenses.values()), "by_category": by_cat}
    def save(self):
        with open(self.path, "w") as f: json.dump([e.to_dict() for e in self.expenses.values()], f, indent=2)
    def load(self):
        if not os.path.exists(self.path): return
        try:
            with open(self.path) as f: data = json.load(f)
            self.expenses = {d["id"]: Expense(**d) for d in data}
            self._next_id = max(self.expenses, default=0) + 1
        except (json.JSONDecodeError, KeyError, TypeError, ExpenseError) as err:
            print("Warning: could not load data file:", err)

# ---- scripted demo ----
if os.path.exists("demo_expenses.json"): os.remove("demo_expenses.json")
t = ExpenseTracker("demo_expenses.json")
t.add("Groceries", 1250.50, "food", "2025-01-05"); t.add("Bus pass", 800, "transport", "2025-01-06"); t.add("Pizza", 450, "Food", "2025-01-07")
t.edit(2, amount=850)
t.delete(3)
for bad in (("", 10), ("Tea", "abc"), ("Tea", -5)):
    try: t.add(*bad)
    except ExpenseError as e: print("Rejected:", e)
print(*[str(e) for e in t.search("")], sep="\n")
print(t.summary())
print("Reloaded from disk:", len(ExpenseTracker("demo_expenses.json").expenses), "expenses")
```
