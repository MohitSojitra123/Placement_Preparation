### Q1. What is Python, and what are its main features?

**Answer:** Python is a high-level, general-purpose, interpreted language created by Guido van Rossum (1991). ### Q2. What is the difference between interpreted and compiled languages?

**Answer:** A compiled language (C, C++) is translated entirely into machine code by a compiler before execution, giving speed but requiring a build step. An interpreted language is executed statement by statement by an interpreter at run time. CPython does both: it compiles source to bytecode, then the Python Virtual Machine interprets that bytecode.

```python
import dis
def add(a, b):
    return a + b
dis.dis(add)  # shows the bytecode CPython compiled before interpreting it
```

### Q3. What is the difference between Python interactive mode and script mode?

**Answer:** Interactive mode (the `>>>` REPL) executes each statement immediately and echoes results, good for experiments. Script mode runs a saved `.py` file from top to bottom, good for real programs; results are only shown via print().

```python
import code
# Simulate a REPL line vs running a script block
console = code.InteractiveConsole()
console.push("2 + 3")          # in interactive mode prints 5 automatically
console.push("print('script-style output')")
exec("x = 2 + 3")               # script style: nothing shown unless printed
```

### Q4. What is indentation in Python, and why is it important?

**Answer:** Indentation (leading whitespace) defines code blocks in Python instead of braces. It is mandatory: all lines in a block must be indented consistently (PEP 8 recommends 4 spaces), otherwise an IndentationError is raised.

```python
for i in range(3):
    print("inside loop", i)
print("outside loop")
try:
    exec("if True:\nprint('bad')")
except IndentationError as e:
    print("IndentationError:", e.msg)
```

### Q10. What is the difference between pass, break, and continue?

### Q16. What is a variable in Python?

**Answer:** A variable is a name that refers to an object stored in memory; it is created by assignment (`x = 10`) and can later refer to a different object.

### Q17. Is Python statically typed or dynamically typed? Explain with an example.

**Answer:** Python is dynamically typed: types belong to objects, not variable names, and are checked at run time, so one name can be rebound to objects of different types. It is still strongly typed (no silent unsafe conversions like `"5" + 5`).

```python
v = 5
print(type(v))
v = "five"
print(type(v))
try:
    print("5" + 5)
except TypeError as e:
    print("TypeError:", e)
```

### Q18. What is the difference between variable declaration and assignment in Python?

**Answer:** Python has no separate declaration: a variable comes into existence at its first assignment (`x = 5`), which binds the name to an object. Using a name never assigned raises NameError. (Annotations like `x: int` declare a type hint only.)

```python
try:
    print(y)
except NameError as e:
    print("NameError:", e)
y = 5          # creation + assignment
z: int         # annotation only, z still unbound
print(y, "z" in globals())
```

### Q19. What is multiple assignment? Write an example.

### Q20. How can you swap two variables without using a third variable?

### Q21. What are mutable and immutable objects?

**Answer:** Mutable objects can be changed in place after creation (list, dict, set, bytearray); immutable objects cannot (int, float, str, tuple, frozenset, bool, bytes) - "changing" them creates a new object.

```python
l = [1, 2]; i0 = id(l); l.append(3); print(id(l) == i0)
s = "ab"; i1 = id(s); s += "c"; print(id(s) == i1)
try:
    "abc"[0] = "x"
except TypeError as e:
    print(e)
```

### Q31. What are arithmetic operators in Python?

**Answer:** Arithmetic operators: `+` addition, `-` subtraction, `*` multiplication, `/` true division, `//` floor division, `%` modulus, `**` exponent.

### Q32. Explain the difference between /, //, and %.

**Answer:** `/` is true division and always returns a float; `//` is floor division (rounds down to the nearest integer); `%` returns the remainder with the sign of the divisor. They satisfy `a == (a // b) * b + a % b`.

### Q33. What is the difference between == and is?

**Answer:** `==` compares values (equality); `is` compares identity (same object in memory). Equal objects need not be identical.

```python
a = [1, 2]; b = [1, 2]; c = a
print(a == b, a is b, a is c)
```

### Q34. Explain comparison operators with examples.

### Q35. What are logical operators? Explain and, or, and not.

**Answer:** `and` is True if both operands are true; `or` is True if at least one is true; `not` inverts a truth value. `and`/`or` return one of their operands, not necessarily a bool.

```python
print(True and False, True or False, not True)
print(0 or "default", 5 and 6, "" and 1)
```

### Q36. What are assignment operators?

### Q38. What is the purpose of the membership operators in and not in?

### Q39. Explain operator precedence in Python.

**Answer:** Precedence decides evaluation order, highest to lowest: parentheses, `**`, unary `+ - ~`, `* / // %`, `+ -`, shifts, `&`, `^`, `|`, comparisons/`in`/`is`, `not`, `and`, `or`, ternary, assignment. `**` is right-associative; others are left-associative.

### Q40. What is the difference between x += 1 and x = x + 1?

```python
a = [1]; b = a
a += [2]; print(a, b)      # in place: b also changed
a = [1]; b = a
a = a + [2]; print(a, b)   # new list: b unchanged
```

### Q44. Explain short-circuit evaluation with an example.

**Answer:** With `and`/`or`, Python stops evaluating as soon as the result is determined: `False and X` never evaluates X, `True or X` never evaluates X. This avoids errors and unnecessary work.

```python
def noisy():
    print("noisy called")
    return True
print(False and noisy())  # noisy not called
print(True or noisy())    # noisy not called
print(True and noisy())   # called
d = {}
print(bool(d) and d["k"])  # safe guard, no KeyError
```

### Q45. Write a program to determine whether a number is divisible by both 3 and 5.

```python
def check(n):
    if n % 3 == 0 and n % 5 == 0:
        return f"{n} is divisible by both 3 and 5"
    return f"{n} is not divisible by both 3 and 5"
for n in (30, 9, 45, 7):
    print(check(n))
```

### Q52. Write a program to find the largest of three numbers.

### Q53. Write a program to determine whether a year is a leap year.

**Answer:** A year is a leap year if divisible by 4, except century years, which must be divisible by 400.

```python
def is_leap(y):
    return y % 4 == 0 and (y % 100 != 0 or y % 400 == 0)
for y in (1900, 2000, 2024, 2023):
    print(y, "leap" if is_leap(y) else "not leap")
```

### Q54. Write a program to check whether a character is a vowel or consonant.

```python
def classify(ch):
    if len(ch) != 1 or not ch.isalpha():
        return "not a letter"
    return "vowel" if ch.lower() in "aeiou" else "consonant"
for c in "aEkZ5":
    print(c, classify(c))
```

### Q55. Write a program to assign grades based on marks.

### Q59. Write a program to check whether three sides can form a valid triangle.

**Answer:** Sides form a valid triangle if all are positive and the sum of any two is greater than the third (triangle inequality).

```python
def valid(a, b, c):
    return a > 0 and b > 0 and c > 0 and a + b > c and a + c > b and b + c > a
for s in [(3, 4, 5), (1, 2, 3), (5, 5, 5), (0, 4, 5)]:
    print(s, "valid" if valid(*s) else "invalid")
```

### Q60. Write a program to calculate an electricity bill using different price slabs.

**Answer:** Apply a different rate to each slab of units (slab/tiered billing). Example slabs: first 100 units at 1.50, next 100 at 2.50, next 200 at 4.00, above 400 at 6.00.

```python
def bill(units):
    slabs = [(100, 1.5), (100, 2.5), (200, 4.0), (float("inf"), 6.0)]
    total, left = 0.0, units
    for size, rate in slabs:
        used = min(left, size)
        total += used * rate
        left -= used
        if left <= 0:
            break
    return total
for u in (50, 150, 350, 500):
    print(u, "units -> bill:", bill(u))
```

## Topic 5: Loops and Iteration (61–75)

### Q61. What is the difference between for and while loops?

**Answer:** A `for` loop iterates over a known sequence/iterable (definite iteration). A `while` loop repeats as long as a condition is true (indefinite iteration) and needs the condition to eventually become false.

```python
for i in range(3):
    print("for", i)
n = 3
while n > 0:
    print("while", n)
    n -= 1
```

### Q62. How does the range() function work?

**Answer:** `range(start, stop, step)` produces a lazy immutable sequence of integers from start up to (excluding) stop, moving by step; start defaults to 0 and step to 1; step may be negative.

```python
print(list(range(5)), list(range(2, 10, 3)), list(range(5, 0, -2)))
```

### Q63. What is the difference between range(5) and range(1, 5)?

**Answer:** `range(5)` gives 0,1,2,3,4 (starts at 0). `range(1, 5)` gives 1,2,3,4 (starts at 1). Both exclude the stop value.

```python
print(list(range(5)))
print(list(range(1, 5)))
```

### Q64. Explain the use of break inside a loop.

**Answer:** `break` immediately exits the nearest enclosing loop and continues after it (the loop's else block is skipped).

```python
for n in [3, 8, 12, 5]:
    if n > 10:
        print("found", n)
        break
    print("checked", n)
```

### Q65. Explain the use of continue inside a loop.

**Answer:** `continue` skips the remaining statements of the current iteration and proceeds to the next iteration.

```python
for i in range(1, 7):
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()
```

### Q66. What is a nested loop?

**Answer:** A nested loop is a loop inside another loop; the inner loop completes all its iterations for each iteration of the outer loop (used for tables, matrices, patterns).

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end="\t")
    print()
```

### Q67. What is an infinite loop, and how can it be stopped?

**Answer:** An infinite loop never terminates because its condition never becomes false (e.g., `while True:` without a break). It can be stopped by a `break`/`return`/exception inside, or externally with Ctrl+C (KeyboardInterrupt).

```python
count = 0
while True:
    count += 1
    if count == 3:
        break   # the exit condition
print("stopped after", count)
```

### Q68. What is the purpose of the else block in a loop?

### Q73. Write a program to calculate the factorial of a number using a loop.

**Answer:** Multiply the numbers 1..n; factorial of 0 is 1 and negative numbers are undefined.

```python
def factorial(n):
    if n < 0:
        raise ValueError("negative")
    f = 1
    for i in range(2, n + 1):
        f *= i
    return f
print(factorial(0), factorial(5), factorial(10))
```

### Q74. Write a program to count the digits in an integer.

**Answer:** Repeatedly integer-divide by 10 until the number becomes 0, counting the steps (0 has one digit).

```python
def count_digits(n):
    n = abs(n)
    if n == 0:
        return 1
    c = 0
    while n > 0:
        n //= 10
        c += 1
    return c
print(count_digits(0), count_digits(12345), count_digits(-9876))
```

### Q75. Write a program to reverse an integer without converting it to a string.

**Answer:** Extract the last digit with `% 10`, append it to the reversed number (`rev = rev*10 + d`) and drop it with `// 10`.

```python
def reverse_int(n):
    sign = -1 if n < 0 else 1
    n, rev = abs(n), 0
    while n > 0:
        rev = rev * 10 + n % 10
        n //= 10
    return sign * rev
print(reverse_int(12345), reverse_int(-120), reverse_int(7))
```

### Q102. Write a program to check whether a string is a palindrome.

### Q103. Write a program to count vowels, consonants, digits, and spaces in a string.

**Answer:** Loop over characters and classify each with `isalpha()`, `isdigit()` and `isspace()`; letters not in "aeiou" are consonants.

```python
def count_chars(s):
    v = c = d = sp = 0
    for ch in s:
        if ch.isalpha():
            if ch.lower() in "aeiou":
                v += 1
            else:
                c += 1
        elif ch.isdigit():
            d += 1
        elif ch.isspace():
            sp += 1
    return v, c, d, sp

v, c, d, sp = count_chars("Hello World 2024 and Python 3")
print("vowels", v, "consonants", c, "digits", d, "spaces", sp)
```

### Q105. Write a program to remove duplicate characters from a string while preserving their first occurrence.

**Answer:** Track seen characters in a set and append a character to the result only the first time it is seen.

```python
def dedupe(s):
    seen = set()
    out = []
    for ch in s:
        if ch not in seen:
            seen.add(ch)
            out.append(ch)
    return "".join(out)

print(dedupe("programming"))
```

### Q116. Write a program to find the largest and smallest elements in a list.

```python
def min_max(lst):
    lo = hi = lst[0]
    for x in lst[1:]:
        if x < lo: lo = x
        if x > hi: hi = x
    return lo, hi

data = [4, 9, -2, 7, 15, 0]
print(min_max(data), (min(data), max(data)))
```

### Q117. Write a program to remove duplicate elements from a list.

```python
def unique(lst):
    seen = set(); out = []
    for x in lst:
        if x not in seen:
            seen.add(x); out.append(x)
    return out

print(unique([1, 2, 2, 3, 1, 4]))
```

### Q118. Write a program to find the second-largest distinct number in a list.

### Q119. Write a program to rotate a list to the right by k positions.

### Q120. Write a program to merge two sorted lists into one sorted list without using an external library.

**Answer:** Use two pointers, always taking the smaller head element, then append whatever remains.

```python
def merge_sorted(a, b):
    i = j = 0; out = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            out.append(a[i]); i += 1
        else:
            out.append(b[j]); j += 1
    out.extend(a[i:]); out.extend(b[j:])
    return out

print(merge_sorted([1, 4, 7], [2, 3, 9, 10]))
```

### Q161. Write a program to find the longest consecutive increasing run in a list.

**Answer:** Scan once, tracking the start of the current strictly increasing run and the best run so far.

```python
def longest_run(lst):
    if not lst:
        return []
    best_start, best_len, start = 0, 1, 0
    for i in range(1, len(lst)):
        if lst[i] <= lst[i - 1]:
            start = i
        if i - start + 1 > best_len:
            best_start, best_len = start, i - start + 1
    return lst[best_start:best_start + best_len]

print(longest_run([1, 3, 5, 2, 4, 6, 8, 9, 1]))
```

### Q162. Write a program to find the maximum sum of a contiguous sublist.

**Answer:** Kadane's algorithm: at each element, either extend the current sum or restart; track the maximum.

```python
def max_subarray(lst):
    best = cur = lst[0]
    for x in lst[1:]:
        cur = max(x, cur + x)
        best = max(best, cur)
    return best

print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))
```

```python
def remove_all(lst, x):
    out = []
    for item in lst:
        if item != x:
            out.append(item)
    return out

def remove_all_inplace(lst, x):
    while x in lst:
        lst.remove(x)
    return lst

print(remove_all([1, 2, 3, 2, 4, 2], 2))
print(remove_all_inplace([1, 2, 3, 2, 4, 2], 2))
```

### Q165. Write a program to rearrange a list so that positive and negative numbers alternate when possible.

**Answer:** Separate positives and negatives, then interleave them; leftovers go at the end.

```python
def alternate(lst):
    pos = [x for x in lst if x >= 0]
    neg = [x for x in lst if x < 0]
    out = []
    i = j = 0
    turn_pos = True
    while i < len(pos) and j < len(neg):
        if turn_pos:
            out.append(pos[i]); i += 1
        else:
            out.append(neg[j]); j += 1
        turn_pos = not turn_pos
    return out + pos[i:] + neg[j:]

print(alternate([1, 2, 3, -1, -2, -3, -4, 5]))
```

### Q199. Write a program to update a nested dictionary without accidentally overwriting unrelated fields.

**Answer:** Update recursively: merge dicts at matching keys instead of replacing the whole sub-dict.

```python
def deep_update(base, new):
    for k, v in new.items():
        if isinstance(v, dict) and isinstance(base.get(k), dict):
            deep_update(base[k], v)
        else:
            base[k] = v
    return base

cfg = {"db": {"host": "localhost", "port": 5432}, "debug": False}
deep_update(cfg, {"db": {"port": 6543}, "debug": True})
print(cfg)
```

### Q200. Write a program to implement a simple phone book with add, search, update, and delete operations.

## Topic 20: Recursion (261–275)

### Q261. What is recursion?

**Answer:** Recursion is when a function calls itself to solve smaller instances of the same problem until reaching a simple case.

```python
def countdown(n):
    if n == 0:
        print("liftoff")
        return
    print(n)
    countdown(n - 1)
countdown(3)
```

### Q262. What is a base case, and why is it necessary?

**Answer:** The base case is the condition where the function returns without recursing. Without it, the function would call itself forever (until the recursion limit is hit), so it guarantees termination.

```python
def fact(n):
    if n <= 1:      # base case
        return 1
    return n * fact(n - 1)
print(fact(5))
```

### Q263. What happens when recursion does not reach a base case?

**Answer:** Each call adds a stack frame, so without a base case Python eventually raises `RecursionError` (maximum recursion depth exceeded) rather than looping forever.

```python
def no_base(n):
    return no_base(n + 1)
try:
    no_base(0)
except RecursionError as e:
    print("RecursionError:", e)
```

### Q264. Write a recursive function to calculate factorial.

**Answer:** factorial(n) = n * factorial(n-1) with base case factorial(0) = 1.

```python
def factorial(n):
    if n < 0:
        raise ValueError("negative")
    return 1 if n <= 1 else n * factorial(n - 1)
print(factorial(0), factorial(5), factorial(10))
```

### Q265. Write a recursive function to calculate the nth Fibonacci number.

**Answer:** fib(n) = fib(n-1) + fib(n-2) with fib(0)=0, fib(1)=1. This plain version is exponential; memoization (`lru_cache`) makes it fast.

```python
from functools import lru_cache
@lru_cache(maxsize=None)
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)
print([fib(i) for i in range(10)], fib(50))
```

### Q266. Write a recursive function to calculate the sum of digits of an integer.

**Answer:** sum_digits(n) = n % 10 + sum_digits(n // 10), base case n == 0.

```python
def sum_digits(n):
    n = abs(n)
    if n == 0:
        return 0
    return n % 10 + sum_digits(n // 10)
print(sum_digits(12345), sum_digits(907))
```

### Q267. Write a recursive function to reverse a string.

**Answer:** reverse(s) = reverse(s[1:]) + s[0], base case: empty or single char.

```python
def rev(s):
    if len(s) <= 1:
        return s
    return rev(s[1:]) + s[0]
print(rev("hello"), rev(""))
```

### Q268. Write a recursive function to check whether a string is a palindrome.

**Answer:** A string is a palindrome if first and last characters match and the middle is also a palindrome; base case is length <= 1.

```python
def is_pal(s):
    s = "".join(c.lower() for c in s if c.isalnum())
    def rec(t):
        if len(t) <= 1:
            return True
        return t[0] == t[-1] and rec(t[1:-1])
    return rec(s)
print(is_pal("racecar"), is_pal("A man, a plan, a canal: Panama"), is_pal("hello"))
```

### Q269. Write a recursive function to calculate the greatest common divisor of two numbers.

**Answer:** Euclid's algorithm: gcd(a, b) = gcd(b, a % b), base case b == 0 returns a.

```python
def gcd(a, b):
    return a if b == 0 else gcd(b, a % b)
print(gcd(48, 18), gcd(17, 5), gcd(0, 9))
```

### Q270. Write a recursive function to calculate a number raised to an integer power.

**Answer:** power(x, n) = x * power(x, n-1); faster by squaring: halve the exponent. Negative exponents use the reciprocal.

```python
def power(x, n):
    if n < 0:
        return 1 / power(x, -n)
    if n == 0:
        return 1
    half = power(x, n // 2)
    return half * half if n % 2 == 0 else half * half * x
print(power(2, 10), power(3, 5), power(2, -2))
```

### Q271. Write a recursive function to flatten a nested list.

**Answer:** If an element is a list, recursively flatten it; otherwise keep it.

```python
def flatten(lst):
    out = []
    for item in lst:
        if isinstance(item, list):
            out.extend(flatten(item))
        else:
            out.append(item)
    return out
print(flatten([1, [2, [3, [4, 5]], 6], [], 7]))
```

### Q272. Explain the difference between recursion and iteration.

**Answer:** Iteration repeats a block using loops with explicit state; recursion solves a problem by calling itself on smaller subproblems, using the call stack for state. Recursion suits trees and divide-and-conquer and is often clearer, but costs more memory and is limited by recursion depth; loops are usually faster and use constant stack.

```python
def fact_rec(n):
    return 1 if n <= 1 else n * fact_rec(n - 1)
def fact_iter(n):
    r = 1
    for i in range(2, n + 1):
        r *= i
    return r
print(fact_rec(6), fact_iter(6))
```

### Q273. What is the recursion limit in Python, and why does it exist?

**Answer:** CPython limits recursion depth (default about 1000) to prevent the C stack from overflowing and crashing the interpreter on runaway recursion. Check with `sys.getrecursionlimit()` and change with `sys.setrecursionlimit()` (carefully).

```python
import sys
print(sys.getrecursionlimit())
def depth(n):
    return depth(n + 1) if n < 10**6 else n
try:
    depth(0)
except RecursionError:
    print("hit the recursion limit")
```

### Q274. Trace the call stack for a recursive factorial function.

**Answer:** For factorial(3), calls stack up: fact(3) -> fact(2) -> fact(1) (base, returns 1); then they unwind: fact(2) = 2*1 = 2, fact(3) = 3*2 = 6. The code prints the trace with indentation by depth.

```python
def fact(n, depth=0):
    pad = "  " * depth
    print(f"{pad}fact({n}) called")
    if n <= 1:
        print(f"{pad}fact({n}) returns 1")
        return 1
    r = n * fact(n - 1, depth + 1)
    print(f"{pad}fact({n}) returns {r}")
    return r
fact(4)
```

### Q275. Rewrite a recursive list-sum function using a loop.

**Answer:** A loop with an accumulator replaces recursion and avoids stack growth.

```python
def rsum(lst):
    return 0 if not lst else lst[0] + rsum(lst[1:])
def lsum(lst):
    total = 0
    for x in lst:
        total += x
    return total
data = [1, 2, 3, 4, 5]
print(rsum(data), lsum(data))
```

### Q278. Explain try, except, else, and finally.

**Answer:** `try` holds risky code; `except` handles an exception; `else` runs only if no exception occurred; `finally` always runs (cleanup).

```python
def div(a, b):
    try:
        r = a / b
    except ZeroDivisionError:
        print("except: division by zero")
    else:
        print("else: success", r)
    finally:
        print("finally: always runs")
div(4, 2)
div(1, 0)
```

### Q279. What is the purpose of raising an exception using raise?

**Answer:** `raise` signals an error deliberately, e.g. when input violates a rule, so callers can handle it instead of continuing with bad data. It can also re-raise the current exception with bare `raise`.

```python
def sqrt_checked(x):
    if x < 0:
        raise ValueError("x must be non-negative")
    return x ** 0.5
try:
    sqrt_checked(-4)
except ValueError as e:
    print("Caught:", e)
```

### Q280. What is a custom exception?

**Answer:** A custom exception is a user-defined class inheriting from `Exception` (or a subclass), used to describe domain-specific errors.

```python
class InsufficientFunds(Exception):
    def __init__(self, balance, amount):
        super().__init__(f"balance {balance} < requested {amount}")
        self.balance, self.amount = balance, amount
try:
    raise InsufficientFunds(50, 80)
except InsufficientFunds as e:
    print("Caught:", e, e.amount)
```

### Q281. Why should specific exceptions generally be caught instead of using a bare except?

**Answer:** A bare `except:` catches everything, including KeyboardInterrupt and SystemExit, and hides bugs. Catching specific exceptions handles only errors you expect and lets unexpected ones surface.

```python
def parse(s):
    try:
        return int(s)
    except ValueError:      # specific
        return None
print(parse("12"), parse("x"))
try:
    parse(None)             # TypeError is not hidden
except TypeError as e:
    print("unexpected error surfaced:", e)
```

### Q282. What is exception propagation?

**Answer:** If an exception is not handled in a function, it propagates up the call stack to the caller, and so on, until a handler catches it or the program terminates.

```python
def c(): return 1 / 0
def b(): return c()
def a():
    try:
        return b()
    except ZeroDivisionError:
        return "handled in a()"
print(a())
```

### Q283. What is exception chaining using raise ... from ...?

**Answer:** `raise NewError(...) from original` sets `__cause__`, chaining the new exception to the original so the traceback shows both and the root cause is kept. `from None` suppresses the context.

```python
def load(s):
    try:
        return int(s)
    except ValueError as e:
        raise RuntimeError("config invalid") from e
try:
    load("abc")
except RuntimeError as e:
    print(e, "| cause:", repr(e.__cause__))
```

### Q284. What is the difference between assert and raise?

**Answer:** `assert cond, msg` is a debugging check that raises AssertionError if cond is false and is removed with `python -O`; use it for internal invariants. `raise` always raises and is for validating input and signalling real error conditions.

```python
try:
    assert 1 + 1 == 3, "math is broken"
except AssertionError as e:
    print("AssertionError:", e)
def set_age(a):
    if a < 0:
        raise ValueError("age cannot be negative")
try:
    set_age(-1)
except ValueError as e:
    print("ValueError:", e)
```

### Q285. Write a program that safely divides two numbers and handles division by zero.

### Q286. Write a program that validates integer input and handles invalid input.

**Answer:** Convert with `int()` and catch `ValueError`. Input is simulated with hard-coded values rather than `input()`.

### Q287. Create a custom exception for an invalid age.

**Answer:** Subclass `Exception` and raise it when the age is out of range.

```python
class InvalidAgeError(Exception):
    """Raised when age is invalid."""
    def __init__(self, age, message="Age must be between 0 and 120"):
        self.age = age
        super().__init__(f"{message} (got {age})")
def set_age(age):
    if not 0 <= age <= 120:
        raise InvalidAgeError(age)
    return age
for a in [30, -3, 150]:
    try:
        print("ok", set_age(a))
    except InvalidAgeError as e:
        print("error:", e)
```

### Q288. Write a program that raises an exception when a list contains no valid values.

**Answer:** Filter the valid values; if none remain, raise `ValueError`.

```python
def average_valid(values):
    valid = [v for v in values if isinstance(v, (int, float)) and v >= 0]
    if not valid:
        raise ValueError("list contains no valid values")
    return sum(valid) / len(valid)
print(average_valid([4, -1, "x", 6]))
try:
    average_valid([-1, "a", None])
except ValueError as e:
    print("Error:", e)
```

### Q289. Explain what happens if an exception occurs inside a finally block.

**Answer:** If a finally block raises, the new exception replaces any in-flight exception (the original is kept as `__context__`), and code after the raise in finally does not run. A `return` in finally also swallows exceptions, so avoid both.

```python
try:
    try:
        1 / 0
    finally:
        raise KeyError("from finally")
except KeyError as e:
    print("Caught:", repr(e))
    print("Context:", repr(e.__context__))
```

### Q290. Write a program that handles multiple exception types without duplicating unnecessary code.

**Answer:** Group exception types in one tuple: `except (ValueError, TypeError) as e`, sharing one handler. Order separate handlers from specific to general.

```python
def to_int(x):
    try:
        return int(x)
    except (ValueError, TypeError) as e:
        print(f"Cannot convert {x!r}: {type(e).__name__}")
        return 0
print(to_int("5"), to_int("five"), to_int(None))
```

### Q292. Explain the modes r, w, a, x, b, and +.

**Answer:** `r` read (default, file must exist); `w` write (truncates or creates); `a` append (writes at end); `x` exclusive create (error if file exists); `b` binary mode; `+` read and write (e.g. `r+`, `w+`). Modes combine, like `rb` or `a+`.

```python
import tempfile, os
with tempfile.TemporaryDirectory() as d:
    p = os.path.join(d, "m.txt")
    with open(p, "x") as f: f.write("one\n")
    try:
        open(p, "x")
    except FileExistsError:
        print("x: file exists error")
    with open(p, "a") as f: f.write("two\n")
    with open(p, "r+") as f:
        print(f.read().split())
        f.write("three\n")
    with open(p, "rb") as f:
        print(f.read())
    with open(p, "w") as f: pass
    print("size after w:", os.path.getsize(p))
```

### Q293. What is the difference between read(), readline(), and readlines()?

### Q294. What is the difference between write() and writelines()?

### Q295. Why is the with statement preferred for file handling?

### Q296. What is a file pointer, and how do seek() and tell() work?

**Answer:** The file pointer is the current read/write position. `tell()` returns it (bytes from the start) and `seek(offset, whence)` moves it (whence 0 = start, 1 = current, 2 = end; non-zero whence with nonzero offsets only in binary mode).

```python
import tempfile, os
with tempfile.TemporaryDirectory() as d:
    p = os.path.join(d, "s.txt")
    with open(p, "w") as f: f.write("0123456789")
    with open(p, "rb") as f:
        print(f.read(3), f.tell())
        f.seek(7)
        print(f.read(), f.tell())
        f.seek(-4, 2)
        print(f.read())
        f.seek(0)
        print(f.read(2))
```

### Q298. Write a program to count lines, words, and characters in a text file.

**Answer:** Read the file once; lines = iterations, words = `split()` counts, characters = length of the text.

```python
import tempfile, os
def count_file(path):
    lines = words = chars = 0
    with open(path) as f:
        for line in f:
            lines += 1
            words += len(line.split())
            chars += len(line)
    return lines, words, chars
with tempfile.TemporaryDirectory() as d:
    p = os.path.join(d, "t.txt")
    with open(p, "w") as f:
        f.write("hello world\nthis is a test\nlast line\n")
    l, w, c = count_file(p)
    print(f"lines={l} words={w} chars={c}")
```

### Q299. Write a program to copy the contents of one text file to another.

**Answer:** Read from the source and write to the destination, line by line (or with `shutil.copyfile`).

```python
import tempfile, os
def copy_file(src, dst):
    with open(src) as fin, open(dst, "w") as fout:
        for line in fin:
            fout.write(line)
with tempfile.TemporaryDirectory() as d:
    a = os.path.join(d, "a.txt")
    b = os.path.join(d, "b.txt")
    with open(a, "w") as f: f.write("line1\nline2\nline3\n")
    copy_file(a, b)
    with open(b) as f: print(f.read(), end="")
    print("identical:", open(a).read() == open(b).read())
```

### Q300. Write a program to find the most frequent word in a text file.

### Q304. What is a context manager, and what are __enter__() and __exit__() used for?

**Answer:** A context manager defines setup and teardown around a `with` block. `__enter__()` runs on entry and its return value is bound by `as`; `__exit__(exc_type, exc, tb)` runs on exit, even on exceptions, and returning True suppresses the exception.

```python
class Demo:
    def __enter__(self):
        print("enter")
        return self
    def __exit__(self, exc_type, exc, tb):
        print("exit", exc_type.__name__ if exc_type else None)
        return True  # suppress the error
with Demo():
    raise ValueError("boom")
print("continued")
```

### Q305. Write a custom context manager that temporarily manages a resource and cleans it up when finished.

**Answer:** Implement `__enter__`/`__exit__` (or use `contextlib.contextmanager`) so the resource is always released, even on errors.

```python
from contextlib import contextmanager

class Resource:
    def __init__(self, name):
        self.name = name
    def __enter__(self):
        print("acquire", self.name)
        return self
    def __exit__(self, *exc):
        print("release", self.name)
        return False

with Resource("db") as r:
    print("using", r.name)
try:
    with Resource("file"):
        raise RuntimeError("fail")
except RuntimeError:
    print("error handled")

@contextmanager
def managed(name):
    print("open", name)
    try:
        yield name
    finally:
        print("close", name)
with managed("socket") as s:
    print("working with", s)
```

### Q306. What is an iterable?

### Q307. What is an iterator?

**Answer:** An iterator is an object with `__iter__()` (returning itself) and `__next__()` (returning the next value or raising `StopIteration`). It remembers its position.

### Q308. What is the difference between an iterable and an iterator?

**Answer:** An iterable can produce iterators (`__iter__`) and can be looped many times; an iterator is a one-shot stream with state that also implements `__next__`. All iterators are iterables, but not vice versa.

```python
lst = [1, 2, 3]
it = iter(lst)
print(list(it), list(it))     # iterator exhausted after one pass
print(list(lst), list(lst))   # iterable reusable
print(hasattr(lst, "__next__"), hasattr(it, "__next__"))
```

### Q309. What do iter() and next() do?

**Answer:** `iter(obj)` returns an iterator from an iterable; `next(it[, default])` fetches the next item, raising `StopIteration` at the end unless a default is given.

### Q310. What is StopIteration?

**Answer:** `StopIteration` is the exception raised by `next()` when an iterator has no more items. `for` loops catch it silently to end the loop.

### Q311. What is a generator function?

**Answer:** A generator function contains `yield`. Calling it returns a generator (an iterator) that runs lazily, pausing at each `yield` and resuming where it left off.

```python
def gen():
    print("start")
    yield 1
    print("middle")
    yield 2
g = gen()
print(type(g).__name__)
print(next(g))
print(next(g))
```

### Q312. What is the difference between return and yield?

**Answer:** `return` ends a function and gives one value; local state is lost. `yield` hands back a value and suspends the function, keeping local state so it can resume and produce many values.

```python
def with_return():
    return [1, 2, 3]
def with_yield():
    for i in (1, 2, 3):
        yield i
print(with_return())
print(with_yield(), list(with_yield()))
```

### Q313. What is a generator expression?

**Answer:** A generator expression looks like a list comprehension with parentheses: `(x*x for x in data)`. It produces items lazily one at a time.

```python
g = (x * x for x in range(5))
print(g)
print(next(g), sum(g))
```

### Q314. Compare a list comprehension with a generator expression.

**Answer:** A list comprehension builds the full list in memory immediately (indexable, reusable); a generator expression is lazy, uses constant memory, and can be consumed only once.

```python
import sys
lc = [x for x in range(10000)]
ge = (x for x in range(10000))
print(sys.getsizeof(lc) > sys.getsizeof(ge))
print(lc[5], sum(ge), sum(ge))
```

### Q315. Write a generator that produces numbers from 1 to n.

**Answer:** Use a loop with `yield` to produce each number on demand.

```python
def upto(n):
    for i in range(1, n + 1):
        yield i
print(list(upto(5)))
```

### Q316. Write a generator that produces Fibonacci numbers.

**Answer:** Keep two running values and yield the next one forever (consumer decides when to stop).

```python
from itertools import islice
def fib():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b
print(list(islice(fib(), 10)))
```

### Q317. Write a generator that reads a file line by line.

**Answer:** Yield each line from an open file so only one line is in memory at a time.

```python
import tempfile, os
def read_lines(path):
    with open(path) as f:
        for line in f:
            yield line.rstrip("\n")
with tempfile.TemporaryDirectory() as d:
    p = os.path.join(d, "x.txt")
    with open(p, "w") as f:
        f.write("a\nb\nc\n")
    for line in read_lines(p):
        print(line)
```

### Q318. Write a custom iterator class that counts down from a starting number.

**Answer:** Implement `__iter__` returning self and `__next__` that returns the next value and raises `StopIteration` at the end.

```python
class CountDown:
    def __init__(self, start):
        self.cur = start
    def __iter__(self):
        return self
    def __next__(self):
        if self.cur < 0:
            raise StopIteration
        v = self.cur
        self.cur -= 1
        return v
print(list(CountDown(5)))
```

### Q319. Explain how generators save memory.

**Answer:** Generators compute and hold one value at a time instead of materialising a whole sequence, so memory stays O(1) regardless of how many items are produced; they also allow infinite streams.

```python
import sys
big_list = [i for i in range(1_000_000)]
big_gen = (i for i in range(1_000_000))
print("list bytes:", sys.getsizeof(big_list))
print("gen bytes:", sys.getsizeof(big_gen))
print(sum(big_gen))
```

### Q320. What happens when you try to iterate over an already exhausted iterator?

**Answer:** An exhausted iterator yields nothing: further `next()` calls raise `StopIteration` and a `for` loop simply runs zero times. It does not reset; create a new iterator to start again.

```python
it = iter([1, 2])
print(list(it))
print(list(it))
for x in it:
    print("never printed")
try:
    next(it)
except StopIteration:
    print("StopIteration")
```

### Q336. Write a program to check whether a number is an Armstrong number.

**Answer:** An Armstrong number equals the sum of its digits each raised to the power of the digit count (153 = 1^3+5^3+3^3).

```python
def is_armstrong(n):
    s = str(n)
    return n == sum(int(c) ** len(s) for c in s)
print([n for n in range(1, 1000) if is_armstrong(n)])
```

### Q337. Write a program to check whether a number is a perfect number.

**Answer:** A perfect number equals the sum of its proper divisors (6 = 1+2+3).

```python
def is_perfect(n):
    return n > 1 and sum(d for d in range(1, n // 2 + 1) if n % d == 0) == n
print([n for n in range(1, 10000) if is_perfect(n)])
```

### Q338. Write a program to generate all prime numbers in a given range.

**Answer:** Use the Sieve of Eratosthenes or trial division up to the square root.

```python
def primes_in_range(lo, hi):
    if hi < 2: return []
    sieve = [True] * (hi + 1)
    sieve[0] = sieve[1] = False
    for i in range(2, int(hi ** 0.5) + 1):
        if sieve[i]:
            for j in range(i * i, hi + 1, i):
                sieve[j] = False
    return [n for n in range(max(lo, 2), hi + 1) if sieve[n]]
print(primes_in_range(10, 50))
```

### Q339. Write a program to find the greatest common divisor and least common multiple of two numbers.

**Answer:** GCD via the Euclidean algorithm; LCM = a*b // gcd.

```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a
def lcm(a, b):
    return a * b // gcd(a, b)
print(gcd(48, 18), lcm(4, 6))
```

### Q340. Write a program to convert a decimal integer to binary without using bin().

**Answer:** Repeatedly divide by 2 and collect remainders, then reverse them.

```python
def to_binary(n):
    if n == 0: return "0"
    neg, n = n < 0, abs(n)
    bits = []
    while n:
        bits.append(str(n % 2))
        n //= 2
    return ("-" if neg else "") + "".join(reversed(bits))
for x in (0, 5, 10, 255):
    print(x, to_binary(x), to_binary(x) == bin(x)[2:])
```

### Q341. Write a program to convert a binary string to a decimal integer without using int(value, 2).

**Answer:** Process digits left to right: `result = result*2 + bit`.

```python
def from_binary(s):
    result = 0
    for ch in s:
        if ch not in "01":
            raise ValueError("not binary: " + s)
        result = result * 2 + (ch == "1")
    return result
for s in ("0", "101", "1010", "11111111"):
    print(s, from_binary(s), int(s, 2))
```

### Q343. Write a program to find the first non-repeating character in a string.

### Q345. Write a program to find all duplicate words in a sentence.

```python
import string
def duplicates(sentence):
    seen, dups = set(), []
    for w in sentence.lower().split():
        w = w.strip(string.punctuation)
        if w in seen and w not in dups:
            dups.append(w)
        seen.add(w)
    return dups
print(duplicates("The cat and the dog and the bird."))
```

### Q346. Write a program to implement a stack using a list.

**Answer:** A stack is LIFO: `append` pushes and `pop()` removes from the end.

```python
class Stack:
    def __init__(self): self._d = []
    def push(self, x): self._d.append(x)
    def pop(self):
        if not self._d: raise IndexError("pop from empty stack")
        return self._d.pop()
    def peek(self): return self._d[-1]
    def is_empty(self): return not self._d
    def __len__(self): return len(self._d)
s = Stack()
for i in (1, 2, 3): s.push(i)
print(s.pop(), s.peek(), len(s))
```

### Q347. Write a program to implement a queue using a list.

**Answer:** A queue is FIFO: enqueue with `append`, dequeue with `pop(0)` (O(n); `collections.deque` is better for performance).

```python
class Queue:
    def __init__(self): self._d = []
    def enqueue(self, x): self._d.append(x)
    def dequeue(self):
        if not self._d: raise IndexError("dequeue from empty queue")
        return self._d.pop(0)
    def is_empty(self): return not self._d
    def __len__(self): return len(self._d)
q = Queue()
for i in "abc": q.enqueue(i)
print(q.dequeue(), q.dequeue(), len(q))
```

### Q348. Write a program to simulate a simple inventory using a dictionary.

```python
inventory = {}
def add(item, qty, price):
    if item in inventory:
        inventory[item]["qty"] += qty
    else:
        inventory[item] = {"qty": qty, "price": price}
def remove(item, qty):
    if item not in inventory or inventory[item]["qty"] < qty:
        print("Cannot remove", qty, item)
        return
    inventory[item]["qty"] -= qty
    if inventory[item]["qty"] == 0:
        del inventory[item]
add("pen", 10, 1.5); add("book", 5, 12.0); add("pen", 5, 1.5)
remove("book", 2); remove("pen", 100)
for k, v in inventory.items():
    print(k, v)
print("Total value:", sum(v["qty"] * v["price"] for v in inventory.values()))
```

### Q358. What is the difference between instance methods, class methods, and static methods?

### Q383. What is the purpose of super()?

### Q384. What is the method resolution order (MRO)?

**Answer:** The MRO is the order in which Python searches classes for an attribute, computed by C3 linearization; view it via `Class.__mro__` or `Class.mro()`. It keeps children before parents and respects the order of the bases.

```python
class A: pass
class B(A): pass
class C(A): pass
class D(B, C): pass
print([c.__name__ for c in D.mro()])
```

### Q385. How does Python resolve methods in multiple inheritance?

**Answer:** Python searches the class and its bases in MRO order (C3 linearization) and uses the first matching attribute: left-to-right across bases, each class before its parents and each class only once.

```python
class X:
    def f(self): return "X"
class Y:
    def f(self): return "Y"
class Z(X, Y): pass
class W(Y, X): pass
print(Z().f(), W().f())
print([c.__name__ for c in Z.__mro__])
```

### Q388. Demonstrate constructor execution across a three-level inheritance hierarchy.

### Q389. Create a multiple-inheritance example and explain its MRO.

**Answer:** Class `Child(Father, Mother)` inherits from both. The MRO is Child, Father, Mother, Person, object, so `Father`'s methods win over `Mother`'s and `Person` comes last.

```python
class Person:
    def trait(self): return "person"
class Father(Person):
    def trait(self): return "father"
class Mother(Person):
    def trait(self): return "mother"
    def cook(self): return "mother cooks"
class Child(Father, Mother): pass
c = Child()
print(c.trait(), c.cook())
print(" -> ".join(k.__name__ for k in Child.__mro__))
```

### Q390. Explain the diamond inheritance problem and how Python's MRO addresses it.

**Answer:** The diamond problem arises when D inherits B and C, both inheriting A: it is ambiguous which version of an A-method (and how many times A's constructor) is used. Python's C3 MRO yields a single consistent order (D, B, C, A) and `super()` visits each class once.

```python
class A:
    def __init__(self): print("A")
class B(A):
    def __init__(self):
        print("B"); super().__init__()
class C(A):
    def __init__(self):
        print("C"); super().__init__()
class D(B, C):
    def __init__(self):
        print("D"); super().__init__()
D()
print([k.__name__ for k in D.__mro__])
```

### Q392. What is the difference between method overriding and method overloading?

**Answer:** Overriding: a subclass redefines a parent's method with the same signature (runtime, across classes). Overloading: several methods with the same name but different parameters in one class (compile-time in languages like Java; not natively supported in Python).

### Q393. Does Python support traditional compile-time method overloading? Explain.

**Answer:** No. A later definition with the same name simply replaces the earlier one. Overloading is emulated with default arguments, `*args`/`**kwargs`, type checks, or `functools.singledispatch`/`singledispatchmethod`.

### Q401. What are magic methods, also called dunder methods?

**Answer:** Magic (dunder) methods are specially named methods like `__init__`, `__add__`, `__len__` that Python calls implicitly in response to syntax and built-ins (`a + b`, `len(x)`, `print(x)`). They let your classes integrate with the language's operators and protocols.

```python
class P:
    def __init__(self, n): self.n = n
    def __add__(self, o): return P(self.n + o.n)
    def __len__(self): return self.n
    def __repr__(self): return f"P({self.n})"
a, b = P(2), P(3)
print(a + b, len(a + b))
print(a.__add__(b))  # what + really calls
```

### Q402. What is the difference between __str__() and __repr__()?

**Answer:** `__str__` returns a readable, user-friendly string (used by `print` and `str()`); `__repr__` returns an unambiguous, developer-oriented string, ideally one that could recreate the object. If `__str__` is missing, Python falls back to `__repr__`.

```python
class C:
    def __init__(self, x): self.x = x
    def __repr__(self): return f"C(x={self.x!r})"
    def __str__(self): return f"C with x = {self.x}"
c = C("hi")
print(str(c)); print(repr(c)); print([c])  # containers use repr
```

### Q403. What is the purpose of __len__()?

**Answer:** `__len__` defines what `len(obj)` returns; it must return a non-negative int. It is also used for truthiness when `__bool__` is not defined (empty means falsy).

```python
class Bag:
    def __init__(self, *items): self.items = list(items)
    def __len__(self): return len(self.items)
b = Bag(1, 2, 3)
print(len(b), bool(b), bool(Bag()))
```

### Q404. How do __eq__() and __ne__() affect comparisons?

**Answer:** `__eq__` defines `==`; `__ne__` defines `!=`. By default `__ne__` inverts `__eq__` (Python 3), so you normally only define `__eq__`. Defining `__eq__` without `__hash__` makes the class unhashable, so define `__hash__` too if instances are used in sets or as dict keys.

```python
class Pt:
    def __init__(self, x, y): self.x, self.y = x, y
    def __eq__(self, o):
        return isinstance(o, Pt) and (self.x, self.y) == (o.x, o.y)
    def __hash__(self): return hash((self.x, self.y))
print(Pt(1, 2) == Pt(1, 2), Pt(1, 2) != Pt(1, 2), Pt(1, 2) != Pt(3, 4))
print(len({Pt(1, 2), Pt(1, 2)}))
```

### Q405. What are __lt__(), __le__(), __gt__(), and __ge__() used for?

**Answer:** They implement the rich comparison operators `<`, `<=`, `>`, `>=`. Defining them makes objects orderable (e.g. for `sorted`, `min`, `max`); `functools.total_ordering` can derive the rest from `__eq__` plus one of them.

```python
from functools import total_ordering
@total_ordering
class Ver:
    def __init__(self, n): self.n = n
    def __eq__(self, o): return self.n == o.n
    def __lt__(self, o): return self.n < o.n
    def __repr__(self): return f"Ver({self.n})"
print(Ver(1) < Ver(2), Ver(3) >= Ver(2), sorted([Ver(3), Ver(1), Ver(2)]))
```

### Q406. What is the purpose of __add__()?

**Answer:** `__add__(self, other)` defines `self + other`. Return `NotImplemented` for unsupported types so Python can try `other.__radd__`. `__radd__` handles the reflected case (e.g. `sum()` starting from 0).

```python
class N:
    def __init__(self, v): self.v = v
    def __add__(self, o):
        if isinstance(o, N): return N(self.v + o.v)
        if isinstance(o, int): return N(self.v + o)
        return NotImplemented
    __radd__ = __add__
    def __repr__(self): return f"N({self.v})"
print(N(1) + N(2), N(1) + 5, 5 + N(1), sum([N(1), N(2)]))
try: N(1) + "x"
except TypeError as e: print("TypeError:", e)
```

### Q407. How do __getitem__() and __setitem__() work?

**Answer:** `obj[key]` calls `__getitem__(key)` and `obj[key] = value` calls `__setitem__(key, value)`. Keys can be ints, slices, or any hashable; raise `IndexError`/`KeyError` appropriately. `__getitem__` alone also makes an object iterable via the legacy protocol.

```python
class Grid:
    def __init__(self): self.d = {}
    def __getitem__(self, k): return self.d.get(k, 0)
    def __setitem__(self, k, v): self.d[k] = v
g = Grid()
g[(0, 1)] = 5
print(g[(0, 1)], g[(9, 9)])
```

### Q408. What is the purpose of __contains__()?

### Q409. What is the difference between __call__() and a normal method?

**Answer:** `__call__` makes the instance itself callable with `obj(...)`, whereas a normal method is invoked by name `obj.method(...)`. It lets objects behave like functions while keeping state.

```python
class Adder:
    def __init__(self, n): self.n = n
    def __call__(self, x): return x + self.n
    def normal(self, x): return x + self.n
a = Adder(10)
print(a(5), a.normal(5), callable(a))
```

### Q410. What are __iter__() and __next__() used for in a custom iterator?

**Answer:** `__iter__` returns an iterator object (often `self`); `__next__` returns the next value or raises `StopIteration` when finished. A `for` loop calls `iter()` once and then `next()` repeatedly.

```python
class Count:
    def __init__(self, n): self.n, self.i = n, 0
    def __iter__(self): return self
    def __next__(self):
        if self.i >= self.n: raise StopIteration
        self.i += 1
        return self.i
print(list(Count(4)))
```

### Q411. Create a Vector class that supports vector addition using +.

### Q416. What is a decorator in Python?

**Answer:** A decorator is a callable that takes a function and returns a (usually wrapped) replacement, applied with `@name` syntax. `@d def f` is equivalent to `f = d(f)`. Used for logging, caching, auth, timing, etc.

### Q419. What is the purpose of functools.wraps in a decorator?

**Answer:** `functools.wraps` copies metadata (`__name__`, `__doc__`, `__module__`, `__wrapped__`, etc.) from the original function onto the wrapper so introspection, help() and debugging still show the original function.

```python
import functools
def plain(f):
    def w(*a, **k): return f(*a, **k)
    return w
def good(f):
    @functools.wraps(f)
    def w(*a, **k): return f(*a, **k)
    return w
@plain
def a(): "doc a"
@good
def b(): "doc b"
print(a.__name__, a.__doc__)
print(b.__name__, b.__doc__)
```

### Q420. Write a decorator that prints a message before and after a function executes.

**Answer:** The wrapper prints before and after calling the function (using try/finally so the "after" also appears on errors).

```python
import functools
def announce(fn):
    @functools.wraps(fn)
    def wrapper(*a, **k):
        print(f"Before {fn.__name__}")
        try:
            return fn(*a, **k)
        finally:
            print(f"After {fn.__name__}")
    return wrapper
@announce
def add(a, b): return a + b
print(add(2, 3))
```

### Q421. Write a decorator that measures a function's execution time using the standard library.

**Answer:** Use `time.perf_counter()` around the call to measure elapsed time.

```python
import time, functools
def timed(fn):
    @functools.wraps(fn)
    def wrapper(*a, **k):
        t = time.perf_counter()
        try:
            return fn(*a, **k)
        finally:
            print(f"{fn.__name__} took {time.perf_counter() - t:.6f}s")
    return wrapper
@timed
def work(n): return sum(range(n))
print(work(100000))
```

### Q422. Write a decorator that validates the types of function arguments.

**Answer:** A decorator factory takes the expected types, then checks each positional argument with `isinstance` before calling.

```python
import functools
def typed(*types):
    def deco(fn):
        @functools.wraps(fn)
        def wrapper(*args):
            for a, t in zip(args, types):
                if not isinstance(a, t):
                    raise TypeError(f"{fn.__name__}: expected {t.__name__}, got {type(a).__name__}")
            return fn(*args)
        return wrapper
    return deco
@typed(int, int)
def add(a, b): return a + b
print(add(1, 2))
try: add(1, "2")
except TypeError as e: print("Error:", e)
```

### Q423. Write a decorator that counts how many times a function is called.

**Answer:** Store the count as an attribute of the wrapper.

```python
import functools
def count_calls(fn):
    @functools.wraps(fn)
    def wrapper(*a, **k):
        wrapper.calls += 1
        return fn(*a, **k)
    wrapper.calls = 0
    return wrapper
@count_calls
def hi(): return "hi"
hi(); hi(); hi()
print("hi called", hi.calls, "times")
```

### Q424. Write a decorator that caches results for repeated calls.

**Answer:** Cache results in a dict keyed by arguments (arguments must be hashable). `functools.lru_cache` is the built-in equivalent.

```python
import functools
def cache(fn):
    store = {}
    @functools.wraps(fn)
    def wrapper(*args):
        if args not in store:
            store[args] = fn(*args)
        else:
            print("cache hit", args)
        return store[args]
    return wrapper
@cache
def slow_square(n):
    print("computing", n)
    return n * n
print(slow_square(4), slow_square(4), slow_square(5))
```

### Q425. Write a decorator that accepts its own arguments and uses them to control a function's behavior.

**Answer:** A decorator with arguments needs three nested levels: factory(args) -> decorator(fn) -> wrapper(*a).

```python
import functools
def repeat(times):
    def deco(fn):
        @functools.wraps(fn)
        def wrapper(*a, **k):
            result = None
            for _ in range(times):
                result = fn(*a, **k)
            return result
        return wrapper
    return deco
@repeat(3)
def say(msg): print(msg)
say("hello")
```

## Topic 32: Closures, Scope, and Late Binding (426–435)

### Q426. What is a closure in Python?

**Answer:** A closure is a nested function that captures and keeps access to variables from its enclosing function's scope even after that function has finished executing.

```python
def outer(x):
    def inner(y): return x + y
    return inner
add5 = outer(5)
print(add5(3), add5.__closure__[0].cell_contents)
```

### Q427. What conditions must be met for a closure to retain an enclosing variable?

**Answer:** There must be a nested function, it must reference a variable from the enclosing function (a free variable), and the enclosing function must return (or otherwise expose) the inner function. Python then stores the variable in a cell. To rebind it, the inner function needs `nonlocal`.

```python
def make():
    count = 0
    def inc():
        nonlocal count
        count += 1
        return count
    return inc
f = make()
print(f(), f(), f.__code__.co_freevars)
```

### Q428. Explain the difference between global and nonlocal in nested functions.

**Answer:** `global` binds a name to the module-level scope; `nonlocal` binds to the nearest enclosing function scope (not global). `nonlocal` requires the name to already exist in an enclosing function.

```python
x = "module"
def outer():
    x = "outer"
    def g():
        global x
        x = "changed global"
    def n():
        nonlocal x
        x = "changed outer"
    n(); print(x)
    g(); print(x)
outer()
print(x)
```

### Q429. Predict the behavior of a nested function that reads a variable from its enclosing function.

**Answer:** The inner function looks up the name in its own scope, then the enclosing scope at call time, so it reads the enclosing value (the current one, even if reassigned after the inner function was defined). Only reading needs no declaration; assigning would create a local.

```python
def outer():
    msg = "hello"
    def inner(): return msg
    msg = "updated"   # inner sees this
    return inner()
print(outer())
```

### Q430. Explain late binding in closures.

**Answer:** Late binding means a closure looks up the free variable's value when the function is called, not when it was defined. All closures sharing one variable therefore see its latest value.

```python
funcs = []
for i in range(3):
    funcs.append(lambda: i)
print([f() for f in funcs])   # [2, 2, 2]
```

### Q431. Why can functions created inside a loop unexpectedly use the same final loop variable?

**Answer:** Because the loop variable is a single name in one scope; every lambda/def created in the loop refers to that same variable, and by the time they are called the loop has finished with the final value. Fix by binding the current value as a default argument or via a factory/`functools.partial`.

### Q438. What is a recursive function's call stack?

**Answer:** Each function call pushes a frame (arguments, locals, return address) onto the call stack. A recursive function stacks one frame per active call until the base case is reached; the frames then unwind as values are returned. Too many frames raises `RecursionError`.

```python
def depth(n):
    if n == 0: return 0
    return 1 + depth(n - 1)
print(depth(10))
import sys
print("limit:", sys.getrecursionlimit())
try: depth(10**6)
except RecursionError: print("RecursionError: stack too deep")
```

### Q439. Explain memoization and implement it using a dictionary.

**Answer:** Memoization stores the results of function calls keyed by their arguments so repeated calls with the same arguments return the stored result instead of recomputing.

```python
def fib(n, memo={}):
    if n in memo: return memo[n]
    if n < 2: return n
    memo[n] = fib(n - 1) + fib(n - 2)
    return memo[n]
print(fib(50))
print([fib(i) for i in range(10)])
```

### Q440. What is the difference between memoization and ordinary caching?

**Answer:** Memoization is a specific kind of caching: it saves the return values of a pure function keyed by its arguments. ### Q443. Write a recursive function to generate all permutations of a list.

**Answer:** Fix each element in turn as the first and recursively permute the remainder.

```python
def permutations(items):
    if len(items) <= 1:
        return [list(items)]
    result = []
    for i, x in enumerate(items):
        for p in permutations(items[:i] + items[i + 1:]):
            result.append([x] + p)
    return result
r = permutations([1, 2, 3])
print(len(r)); print(r)
```

### Q444. Write a recursive function to generate all subsets of a set of distinct elements.

**Answer:** For each element, subsets either exclude or include it; recurse on the rest.

```python
def subsets(items):
    if not items:
        return [[]]
    rest = subsets(items[1:])
    return rest + [[items[0]] + s for s in rest]
s = subsets([1, 2, 3])
print(len(s)); print(s)
```

### Q445. Explain the limitations of deeply recursive algorithms and propose an iterative alternative.

**Answer:** Deep recursion is limited by the interpreter's recursion limit (default ~1000) and uses stack memory per call, risking `RecursionError`; Python also does no tail-call optimization. Alternatives: rewrite iteratively with a loop or an explicit stack, or use memoization/dynamic programming.

```python
import sys
def sum_rec(n): return 0 if n == 0 else n + sum_rec(n - 1)
def sum_iter(n):
    total = 0
    while n:
        total += n; n -= 1
    return total
print(sum_iter(100000))
try: sum_rec(100000)
except RecursionError: print("recursive version fails at", sys.getrecursionlimit())
# explicit stack: iterative DFS
def dfs(graph, start):
    seen, stack = [], [start]
    while stack:
        n = stack.pop()
        if n not in seen:
            seen.append(n); stack.extend(reversed(graph[n]))
    return seen
print(dfs({1: [2, 3], 2: [4], 3: [], 4: []}, 1))
```

### Q466. Design a custom exception hierarchy for an online shopping application.

**Answer:** Define a base exception for the app and derive specific errors, so callers can catch broadly or narrowly.

```python
class ShopError(Exception):
    """Base for all shop errors."""
class OutOfStockError(ShopError):
    def __init__(self, item, requested, available):
        super().__init__(f"{item}: requested {requested}, only {available} left")
        self.item = item
class PaymentError(ShopError): pass
class CardDeclinedError(PaymentError): pass
class InsufficientFundsError(PaymentError): pass
class CartError(ShopError): pass
class EmptyCartError(CartError): pass

def checkout(cart, stock, balance):
    if not cart: raise EmptyCartError("cart is empty")
    total = 0
    for item, (qty, price) in cart.items():
        if stock.get(item, 0) < qty: raise OutOfStockError(item, qty, stock.get(item, 0))
        total += qty * price
    if total > balance: raise InsufficientFundsError(f"need {total}, have {balance}")
    return total
for cart, bal in [({}, 50), ({"pen": (5, 2)}, 50), ({"pen": (1, 2)}, 1), ({"pen": (2, 2)}, 50)]:
    try: print("paid", checkout(cart, {"pen": 3}, bal))
    except OutOfStockError as e: print("stock:", e)
    except PaymentError as e: print("payment:", type(e).__name__, e)
    except ShopError as e: print("shop:", e)
```

### Q467. Explain exception chaining and write an example using raise ... from ....

**Answer:** Exception chaining links a new exception to the one that caused it. `raise X from Y` sets `X.__cause__ = Y` (explicit); raising inside an `except` block sets `__context__` implicitly. `from None` suppresses the chain. The traceback shows both.

```python
class ConfigError(Exception): pass
def load(value):
    try:
        return int(value)
    except ValueError as e:
        raise ConfigError(f"bad config value {value!r}") from e
try: load("abc")
except ConfigError as e:
    print(e); print("cause:", repr(e.__cause__))
```

### Q471. Implement a custom context manager using a class.

**Answer:** A class-based context manager implements `__enter__` (setup, returns the `as` value) and `__exit__` (cleanup; return True to suppress the exception).

```python
import time
class Timer:
    def __enter__(self):
        self.start = time.perf_counter(); print("start"); return self
    def __exit__(self, exc_type, exc, tb):
        self.elapsed = time.perf_counter() - self.start
        print("exit, exception:", exc_type.__name__ if exc_type else None)
        return False   # do not suppress
with Timer() as t:
    sum(range(1000))
print(t.elapsed >= 0)
try:
    with Timer(): 1 / 0
except ZeroDivisionError: print("error propagated")
```

### Q472. Implement a context manager using contextlib.contextmanager.

### Q473. Create a context manager that temporarily changes a setting and restores its original value afterward.

### Q477. Write a generator that yields prime numbers indefinitely.

**Answer:** Yield each number not divisible by any previously found prime.

```python
from itertools import islice
def primes():
    found = []
    n = 2
    while True:
        if all(n % p for p in found if p * p <= n):
            found.append(n)
            yield n
        n += 1
print(list(islice(primes(), 15)))
```

### Q478. Write a generator that reads a sequence in batches of a given size.

```python
from itertools import islice
def batches(seq, size):
    if size < 1: raise ValueError("size must be >= 1")
    it = iter(seq)
    while chunk := list(islice(it, size)):
        yield chunk
print(list(batches(range(10), 4)))
print(list(batches("abcde", 2)))
```

### Q479. Explain the difference between generator exhaustion and generator reset.

### Q481. Write a generator that delegates part of its work to another generator using yield from.

**Answer:** The outer generator delegates to inner generators with `yield from`.

```python
def countdown(n):
    while n > 0:
        yield n; n -= 1
def letters(s):
    for c in s: yield c
def combined():
    yield from countdown(3)
    yield from letters("ab")
    yield "end"
print(list(combined()))
```

### Q482. Explain lazy evaluation and compare it with eager evaluation.

**Answer:** Lazy evaluation computes values only when needed (generators, `range`, `map`), saving memory and allowing infinite sequences; eager evaluation computes everything up front (lists) which allows reuse and random access but uses more memory/time.

### Q496. Library Management System: Design classes for books, members, and loans. Implement borrowing, returning, availability checks, and validation using OOP.

**Answer:** A library system with `Book`, `Member`, `Loan` and `Library` classes, validation via custom exceptions, borrowing limits, due dates, and availability checks.

```python
from datetime import date, timedelta

class LibraryError(Exception): pass
class BookNotAvailable(LibraryError): pass
class LimitExceeded(LibraryError): pass
class NotFound(LibraryError): pass

class Book:
    def __init__(self, isbn, title, author, copies=1):
        if not isbn or not title: raise ValueError("isbn and title required")
        if copies < 1: raise ValueError("copies must be >= 1")
        self.isbn, self.title, self.author = isbn, title, author
        self.copies, self.available = copies, copies
    def __str__(self): return f"{self.title} by {self.author} [{self.available}/{self.copies}]"

class Member:
    MAX_LOANS = 2
    def __init__(self, member_id, name):
        if not name.strip(): raise ValueError("name required")
        self.member_id, self.name, self.loans = member_id, name, []
    def can_borrow(self): return len(self.loans) < self.MAX_LOANS

class Loan:
    PERIOD = 14
    def __init__(self, book, member, start):
        self.book, self.member, self.start = book, member, start
        self.due = start + timedelta(days=self.PERIOD)
        self.returned = None
    def fine(self, today):
        end = self.returned or today
        return max(0, (end - self.due).days) * 0.5

class Library:
    def __init__(self):
        self.books, self.members, self.history = {}, {}, []
    def add_book(self, book):
        if book.isbn in self.books: self.books[book.isbn].copies += book.copies; self.books[book.isbn].available += book.copies
        else: self.books[book.isbn] = book
    def add_member(self, m): self.members[m.member_id] = m
    def _get(self, d, k, what):
        if k not in d: raise NotFound(f"{what} {k} not found")
        return d[k]
    def is_available(self, isbn): return self._get(self.books, isbn, "book").available > 0
    def borrow(self, member_id, isbn, today):
        m, b = self._get(self.members, member_id, "member"), self._get(self.books, isbn, "book")
        if not m.can_borrow(): raise LimitExceeded(f"{m.name} reached the limit of {m.MAX_LOANS}")
        if any(l.book is b for l in m.loans): raise LibraryError(f"{m.name} already has {b.title}")
        if b.available == 0: raise BookNotAvailable(f"{b.title} is not available")
        b.available -= 1
        loan = Loan(b, m, today); m.loans.append(loan); self.history.append(loan)
        return loan
    def give_back(self, member_id, isbn, today):
        m = self._get(self.members, member_id, "member")
        loan = next((l for l in m.loans if l.book.isbn == isbn), None)
        if loan is None: raise NotFound("no such active loan")
        loan.returned = today; loan.book.available += 1; m.loans.remove(loan)
        return loan.fine(today)

lib = Library()
lib.add_book(Book("1", "Dune", "Herbert")); lib.add_book(Book("2", "Emma", "Austen", 2)); lib.add_book(Book("3", "Ulysses", "Joyce"))
lib.add_member(Member(1, "Ann")); lib.add_member(Member(2, "Bob"))
d0 = date(2025, 1, 1)
print(lib.borrow(1, "1", d0).due)
for args in [(2, "1"), (1, "2"), (1, "3"), (9, "1"), (1, "2")]:
    try: lib.borrow(*args, d0); print("borrowed", args)
    except LibraryError as e: print(type(e).__name__ + ":", e)
print("available Dune?", lib.is_available("1"))
print("fine:", lib.give_back(1, "1", d0 + timedelta(days=20)))
print("available Dune?", lib.is_available("1"))
for b in lib.books.values(): print(b)
```

### Q497. Banking System: Build an account hierarchy with savings and current accounts, transaction history, deposit, withdrawal, transfer, and custom exceptions.

**Answer:** A `Account` base class with `SavingsAccount` (interest, no overdraft) and `CurrentAccount` (overdraft limit), transaction history, transfers and custom exceptions.

```python
from datetime import datetime

class BankError(Exception): pass
class InsufficientFunds(BankError): pass
class InvalidAmount(BankError): pass
class AccountNotFound(BankError): pass

class Transaction:
    def __init__(self, kind, amount, balance, note=""):
        self.kind, self.amount, self.balance, self.note = kind, amount, balance, note
    def __str__(self): return f"{self.kind:<13}{self.amount:>10.2f}  balance={self.balance:.2f} {self.note}"

class Account:
    _next = 1000
    def __init__(self, owner, balance=0.0):
        self.owner, self.balance, self.history = owner, 0.0, []
        self.number = Account._next; Account._next += 1
        if balance: self.deposit(balance)
    @staticmethod
    def _check(amount):
        if not isinstance(amount, (int, float)) or amount <= 0: raise InvalidAmount(f"invalid amount: {amount}")
    def available(self): return self.balance
    def deposit(self, amount):
        self._check(amount); self.balance += amount
        self.history.append(Transaction("deposit", amount, self.balance))
    def withdraw(self, amount, kind="withdraw", note=""):
        self._check(amount)
        if amount > self.available(): raise InsufficientFunds(f"{self.number}: need {amount}, available {self.available()}")
        self.balance -= amount
        self.history.append(Transaction(kind, amount, self.balance, note))
    def __str__(self): return f"{type(self).__name__}#{self.number} {self.owner}: {self.balance:.2f}"

class SavingsAccount(Account):
    def __init__(self, owner, balance=0.0, rate=0.03):
        super().__init__(owner, balance); self.rate = rate
    def add_interest(self):
        i = round(self.balance * self.rate, 2)
        if i > 0: self.deposit(i); self.history[-1].kind = "interest"

class CurrentAccount(Account):
    def __init__(self, owner, balance=0.0, overdraft=500):
        super().__init__(owner, balance); self.overdraft = overdraft
    def available(self): return self.balance + self.overdraft

class Bank:
    def __init__(self): self.accounts = {}
    def open(self, acc): self.accounts[acc.number] = acc; return acc
    def get(self, n):
        if n not in self.accounts: raise AccountNotFound(n)
        return self.accounts[n]
    def transfer(self, src, dst, amount):
        a, b = self.get(src), self.get(dst)
        a.withdraw(amount, "transfer out", f"to {dst}")
        b.deposit(amount); b.history[-1].kind = "transfer in"; b.history[-1].note = f"from {src}"

bank = Bank()
s = bank.open(SavingsAccount("Ann", 1000)); c = bank.open(CurrentAccount("Bob", 100))
bank.transfer(s.number, c.number, 250)
c.withdraw(700)                  # uses overdraft
s.add_interest()
for bad in [lambda: s.withdraw(5000), lambda: c.deposit(-5), lambda: bank.transfer(s.number, 1, 5), lambda: c.withdraw(1000)]:
    try: bad()
    except BankError as e: print(type(e).__name__ + ":", e)
for acc in (s, c):
    print(acc)
    for t in acc.history: print("  ", t)
```

### Q498. Student Result Management: Store student records using dictionaries and lists. Calculate subject averages, rank students, identify ties, and generate formatted reports.

**Answer:** Students are stored as dicts of subject->mark; helper functions compute subject averages, ranking with ties (competition ranking 1,2,2,4), and a formatted report.

```python
students = {
    "Alice": {"math": 90, "sci": 85, "eng": 80},
    "Bob": {"math": 70, "sci": 95, "eng": 75},
    "Cara": {"math": 85, "sci": 80, "eng": 90},
    "Dan": {"math": 60, "sci": 70, "eng": 65},
    "Eve": {"math": 85, "sci": 90, "eng": 80},
}
def validate(students):
    for n, marks in students.items():
        for s, m in marks.items():
            if not 0 <= m <= 100: raise ValueError(f"{n}/{s}: mark {m} out of range")
def subject_averages(students):
    subjects = sorted({s for m in students.values() for s in m})
    return {s: sum(m[s] for m in students.values() if s in m) / sum(1 for m in students.values() if s in m) for s in subjects}
def totals(students): return {n: sum(m.values()) for n, m in students.items()}
def rank(students):
    t = totals(students)
    ordered = sorted(t.items(), key=lambda kv: (-kv[1], kv[0]))
    ranked, prev, pos = [], None, 0
    for i, (n, tot) in enumerate(ordered, 1):
        if tot != prev: pos = i
        ranked.append((pos, n, tot)); prev = tot
    return ranked
def ties(ranked):
    from collections import defaultdict
    g = defaultdict(list)
    for pos, n, _ in ranked: g[pos].append(n)
    return {p: names for p, names in g.items() if len(names) > 1}
def report(students):
    validate(students)
    lines = [f"{'Rank':<5}{'Name':<8}" + "".join(f"{s:>6}" for s in sorted(next(iter(students.values())))) + f"{'Total':>7}{'Avg':>7}", "-" * 40]
    for pos, n, tot in rank(students):
        lines.append(f"{pos:<5}{n:<8}" + "".join(f"{students[n][s]:>6}" for s in sorted(students[n])) + f"{tot:>7}{tot / len(students[n]):>7.1f}")
    lines.append("-" * 40)
    lines.append(f"{'Avg':<13}" + "".join(f"{v:>6.1f}" for v in subject_averages(students).values()))
    return "\n".join(lines)
print(report(students))
print("Ties:", ties(rank(students)))
try: validate({"X": {"math": 120}})
except ValueError as e: print("Validation:", e)
```

### Q499. Custom Data Structures: Implement a stack, queue, and singly linked list using classes, including insertion, deletion, searching, traversal, and edge-case handling.

**Answer:** A `Stack` (list-based, LIFO), `Queue` (deque-based, FIFO) and `LinkedList` (singly linked with insert, delete, search, traversal), each raising clear errors on empty/invalid operations.

```python
from collections import deque

class Stack:
    def __init__(self): self._d = []
    def push(self, x): self._d.append(x)
    def pop(self):
        if not self._d: raise IndexError("pop from empty stack")
        return self._d.pop()
    def peek(self):
        if not self._d: raise IndexError("peek at empty stack")
        return self._d[-1]
    def is_empty(self): return not self._d
    def __len__(self): return len(self._d)
    def __repr__(self): return f"Stack({self._d})"

class Queue:
    def __init__(self): self._d = deque()
    def enqueue(self, x): self._d.append(x)
    def dequeue(self):
        if not self._d: raise IndexError("dequeue from empty queue")
        return self._d.popleft()
    def front(self):
        if not self._d: raise IndexError("front of empty queue")
        return self._d[0]
    def is_empty(self): return not self._d
    def __len__(self): return len(self._d)
    def __repr__(self): return f"Queue({list(self._d)})"

class Node:
    def __init__(self, data, next=None): self.data, self.next = data, next

class LinkedList:
    def __init__(self): self.head, self.size = None, 0
    def append(self, x):
        n = Node(x)
        if not self.head: self.head = n
        else:
            cur = self.head
            while cur.next: cur = cur.next
            cur.next = n
        self.size += 1
    def prepend(self, x): self.head = Node(x, self.head); self.size += 1
    def insert(self, index, x):
        if not 0 <= index <= self.size: raise IndexError("index out of range")
        if index == 0: return self.prepend(x)
        cur = self.head
        for _ in range(index - 1): cur = cur.next
        cur.next = Node(x, cur.next); self.size += 1
    def delete(self, x):
        cur, prev = self.head, None
        while cur:
            if cur.data == x:
                if prev: prev.next = cur.next
                else: self.head = cur.next
                self.size -= 1
                return True
            prev, cur = cur, cur.next
        return False
    def search(self, x):
        for i, v in enumerate(self):
            if v == x: return i
        return -1
    def reverse(self):
        prev, cur = None, self.head
        while cur: cur.next, prev, cur = prev, cur, cur.next
        self.head = prev
    def __iter__(self):
        cur = self.head
        while cur: yield cur.data; cur = cur.next
    def __len__(self): return self.size
    def __repr__(self): return " -> ".join(map(str, self)) + " -> None" if self.head else "empty"

s = Stack()
for i in (1, 2, 3): s.push(i)
print(s, s.pop(), s.peek(), len(s))
q = Queue()
for i in "abc": q.enqueue(i)
print(q, q.dequeue(), q.front())
ll = LinkedList()
for i in (10, 20, 30): ll.append(i)
ll.prepend(5); ll.insert(2, 15)
print(ll, "len", len(ll), "search 20 ->", ll.search(20), "search 99 ->", ll.search(99))
print(ll.delete(5), ll.delete(30), ll.delete(99), ll)
ll.reverse(); print(ll)
for obj, op in [(Stack(), "pop"), (Queue(), "dequeue"), (LinkedList(), "insert")]:
    try: getattr(obj, op)(3, 5) if op == "insert" else getattr(obj, op)()
    except IndexError as e: print("IndexError:", e)
print(LinkedList().delete(1), LinkedList().search(1))
```

### Q500. Mini Python Project — Expense Tracker: Build a command-line expense tracker using only Core Python. Support adding, editing, deleting, searching, categorizing, and summarizing expenses; save and load data from a file; validate input; and use OOP, comprehensions, and exception handling.

**Answer:** An `Expense` class, an `ExpenseTracker` (add/edit/delete/search/summarize, JSON persistence) and a `CLI` class whose command loop is fed scripted commands instead of `input()`. Validation uses exceptions; summaries use comprehensions.

```python
import json, os, tempfile
from datetime import date

class ExpenseError(Exception): pass

class Expense:
    def __init__(self, eid, amount, category, description, when):
        self.id, self.amount = eid, self._valid_amount(amount)
        self.category, self.description = category.strip().lower(), description.strip()
        self.when = self._valid_date(when)
        if not self.category: raise ExpenseError("category required")
    @staticmethod
    def _valid_amount(a):
        try: a = round(float(a), 2)
        except (TypeError, ValueError): raise ExpenseError(f"invalid amount: {a!r}")
        if a <= 0: raise ExpenseError("amount must be positive")
        return a
    @staticmethod
    def _valid_date(d):
        try: return date.fromisoformat(str(d)).isoformat()
        except ValueError: raise ExpenseError(f"invalid date (use YYYY-MM-DD): {d!r}")
    def to_dict(self): return dict(id=self.id, amount=self.amount, category=self.category, description=self.description, when=self.when)
    def __str__(self): return f"#{self.id:<3}{self.when}  {self.category:<10}{self.amount:>9.2f}  {self.description}"

class ExpenseTracker:
    def __init__(self): self.expenses, self._next = {}, 1
    def add(self, amount, category, description, when=None):
        e = Expense(self._next, amount, category, description, when or date.today().isoformat())
        self.expenses[e.id] = e; self._next += 1
        return e
    def _get(self, eid):
        try: return self.expenses[int(eid)]
        except (KeyError, ValueError): raise ExpenseError(f"no expense with id {eid!r}")
    def edit(self, eid, **changes):
        e = self._get(eid)
        d = e.to_dict(); d.update({k: v for k, v in changes.items() if v is not None})
        self.expenses[e.id] = Expense(d["id"], d["amount"], d["category"], d["description"], d["when"])
        return self.expenses[e.id]
    def delete(self, eid): return self.expenses.pop(self._get(eid).id)
    def search(self, text):
        t = text.lower()
        return [e for e in self.expenses.values() if t in e.description.lower() or t in e.category]
    def by_category(self, cat): return [e for e in self.expenses.values() if e.category == cat.lower()]
    def summary(self):
        cats = {e.category for e in self.expenses.values()}
        return {c: round(sum(e.amount for e in self.expenses.values() if e.category == c), 2) for c in sorted(cats)}
    def total(self): return round(sum(e.amount for e in self.expenses.values()), 2)
    def save(self, path):
        with open(path, "w") as f:
            json.dump([e.to_dict() for e in self.expenses.values()], f, indent=2)
    def load(self, path):
        try:
            with open(path) as f: rows = json.load(f)
            loaded = {r["id"]: Expense(r["id"], r["amount"], r["category"], r["description"], r["when"]) for r in rows}
        except (OSError, json.JSONDecodeError, KeyError, TypeError) as e:
            raise ExpenseError(f"could not load {path}: {e}")
        self.expenses = loaded; self._next = max(loaded, default=0) + 1

class CLI:
    HELP = "commands: add <amt> <cat> <desc> | edit <id> <amt> | del <id> | list | search <text> | cat <name> | summary | save | load | quit"
    def __init__(self, tracker, path): self.t, self.path = tracker, path
    def run(self, commands):
        print(self.HELP)
        for line in commands:
            print(f"> {line}")
            parts = line.split()
            if not parts: continue
            cmd, args = parts[0].lower(), parts[1:]
            try:
                if cmd == "quit": print("Bye!"); break
                self.dispatch(cmd, args)
            except ExpenseError as e: print("Error:", e)
            except (IndexError, ValueError): print("Error: bad arguments.", self.HELP)
    def dispatch(self, cmd, a):
        t = self.t
        if cmd == "add": print("Added:", t.add(a[0], a[1], " ".join(a[2:]) or "-", None if len(a) < 1 else "2025-03-01"))
        elif cmd == "edit": print("Edited:", t.edit(a[0], amount=a[1]))
        elif cmd == "del": print("Deleted:", t.delete(a[0]))
        elif cmd == "list": [print(" ", e) for e in t.expenses.values()] or None
        elif cmd == "search": [print(" ", e) for e in t.search(" ".join(a))]
        elif cmd == "cat": [print(" ", e) for e in t.by_category(a[0])]
        elif cmd == "summary":
            for c, v in t.summary().items(): print(f"  {c:<10}{v:>9.2f}")
            print(f"  {'TOTAL':<10}{t.total():>9.2f}")
        elif cmd == "save": t.save(self.path); print("Saved to file")
        elif cmd == "load": t.load(self.path); print("Loaded", len(t.expenses), "expenses")
        else: raise ExpenseError(f"unknown command {cmd!r}")

d = tempfile.mkdtemp(); path = os.path.join(d, "expenses.json")
try:
    cli = CLI(ExpenseTracker(), path)
    cli.run(["add 12.5 food lunch with team", "add 40 transport train ticket", "add abc food oops", "add 80 food groceries",
             "edit 1 15", "list", "search train", "cat food", "summary", "del 99", "save", "del 2", "load", "list", "bogus", "quit", "list"])
finally:
    if os.path.exists(path): os.remove(path)
    os.rmdir(d)
```
